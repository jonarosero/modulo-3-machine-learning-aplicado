## Bloque 2. Carga del dataset

El archivo ya está cargado y sus columnas numéricas convertidas (las comas decimales se cambian por puntos). **No se eliminan filas todavía**: esa decisión es parte de su trabajo en el bloque 3.

```python
DATA_DIR = Path('datos')

df = pd.read_csv(DATA_DIR / 'countries of the world.csv')
df.head()
```

```python
df.columns = df.columns.str.strip()
df['Country'] = df['Country'].str.strip()
df['Region'] = df['Region'].str.strip()

# Convertir a número todas las columnas que no son texto.
for col in df.columns:
    if col not in ['Country', 'Region']:
        df[col] = pd.to_numeric(df[col].astype(str).str.replace(',', '.', regex=False), errors='coerce')

print('Filas:', len(df))
print('Valores faltantes por columna:')
print(df.isna().sum())
```
