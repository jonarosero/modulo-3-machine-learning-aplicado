# 1. Python desde cero

## 1.1 Variables y tipos de datos

Una **variable** es un nombre que apunta a un valor guardado en memoria. En
Python se crea con `=` (asignación), no con un signo de igualdad matemático.

````python
# Creamos variables de distintos tipos.
nombre_tienda = "Ferretería El Tornillo"   # str -> texto
dias_abierto = 6                           # int -> número entero
ingreso_promedio = 245.50                  # float -> número decimal
tiene_sucursal = False                     # bool -> verdadero/falso
telefono_sucursal = None                   # None -> "todavía no existe un valor"

print(nombre_tienda)
print(dias_abierto)
print(ingreso_promedio)
print(tiene_sucursal)
print(telefono_sucursal)
````

`type()` te dice qué tipo de dato es una variable. Es la primera función que
deberías usar cuando algo no se comporta como esperabas.

````python
print(type(nombre_tienda))
print(type(dias_abierto))
print(type(ingreso_promedio))
print(type(tiene_sucursal))
print(type(telefono_sucursal))
````

> `None` **no es lo mismo que 0, que `False` ni que una cadena vacía `""`**.
> `None` significa "ausencia de valor" — lo volverás a ver cuando hablemos de
> datos faltantes en Pandas (Sección 7.7). Es el mismo concepto.

## 1.2 Operadores

### Operadores aritméticos

````python
a, b = 17, 5

print("suma:        ", a + b)
print("resta:       ", a - b)
print("multiplicacion:", a * b)
print("division:    ", a / b)   # division "normal": siempre devuelve float
print("division entera:", a // b)  # descarta la parte decimal
print("modulo (resto):", a % b)    # lo que sobra de la division entera
print("potencia:    ", a ** 2)
````

> `/` y `//` responden preguntas distintas: `17 / 5 = 3.4` (el resultado
> exacto), mientras que `17 // 5 = 3` (cuántas veces cabe 5 en 17, sin
> decimales). El `%` (módulo) es el resto de esa división: `17 % 5 = 2`
> porque `5 × 3 = 15` y sobran `2`.

### Operadores de comparación y lógicos

Una comparación siempre devuelve un `bool` (`True` o `False`).

````python
unidades_vendidas = 12
stock_minimo = 10

print(unidades_vendidas > stock_minimo)   # mayor que
print(unidades_vendidas == stock_minimo)  # igual a (OJO: == compara, = asigna)
print(unidades_vendidas != stock_minimo)  # distinto de

# Operadores logicos: combinan condiciones booleanas.
hay_stock = unidades_vendidas > 0
es_fin_de_semana = True
print(hay_stock and es_fin_de_semana)  # True solo si AMBAS son True
print(hay_stock or es_fin_de_semana)   # True si AL MENOS UNA es True
print(not hay_stock)                   # invierte el valor
````

> El error más común de quien empieza: confundir `=` (asignar un valor) con
> `==` (preguntar si dos valores son iguales). `edad = 18` le pone 18 a
> `edad`; `edad == 18` pregunta "¿`edad` vale 18?" y responde `True`/`False`.

*Practica tú*: antes de seguir, predice en tu cabeza el resultado de
`(5 > 3) and (2 == 2)` y de `(5 > 3) or (1 > 2)` — luego pruébalo en una
celda nueva.

## 1.3 Conversión de tipos

A veces un número llega como texto (por ejemplo, leído de un archivo) y
necesitas convertirlo para poder operar con él.

````python
precio_texto = "15.50"       # esto es un str, no un numero
precio_numero = float(precio_texto)
print(precio_numero, type(precio_numero))

cantidad_texto = "8"
cantidad_numero = int(cantidad_texto)
print(cantidad_numero, type(cantidad_numero))

# Tambien se puede convertir al reves, numero -> texto:
total = precio_numero * cantidad_numero
print("El total es: " + str(total))   # sin str(), esto daria TypeError
````

Para ver un error de verdad (y aprender a leerlo), probemos convertir un

````python
try:
    numero_invalido = int("quince")
except ValueError as error:
    print("Ocurrió un ValueError:", error)
````

`int("quince")` falla porque Python no sabe convertir la palabra "quince" en
un número — solo convierte dígitos (`"15"` sí funciona). Usamos `try/except`
para **atrapar** el error y seguir ejecutando en vez de detener el notebook;
volveremos a `try/except` solo cuando haga falta, no es un tema central del
curso.

> Lee el mensaje de error de adentro hacia afuera: `ValueError` es el
> **tipo** de error, y el texto después de los dos puntos es la causa
> exacta. Acostúmbrate a leer esa última línea primero.

## 1.4 Cadenas de texto (strings)

Un `str` se puede indexar y recortar igual que una lista (lo veremos formal
en 2.1): cada carácter tiene una posición, empezando en 0.

````python
region = "Pichincha"

print(region[0])      # primer caracter
print(region[-1])     # ultimo caracter
print(region[0:4])    # "slicing": de la posicion 0 a la 4 (sin incluirla)
print(len(region))    # cuantos caracteres tiene
print(region.upper())
print(region.lower())
````

Las **f-strings** (`f"..."`) son la forma moderna de construir texto que
combina literales con valores de variables — las verás en casi todo el curso.

````python
producto = "camiseta"
precio = 15.5
unidades = 12

mensaje = f"Se vendieron {unidades} unidades de {producto} a ${precio} c/u"
print(mensaje)

# Tambien se pueden formatear numeros dentro de la f-string:
total = precio * unidades
print(f"Total: ${total:.2f}")   # .2f = 2 decimales, como dinero
````

# 2. Estructuras de datos nativas

Una sola variable guarda un solo valor. Para guardar **varios valores
relacionados** — como las ventas de toda una semana — necesitamos
estructuras de datos. Python trae cuatro básicas, y cada una responde a una
pregunta distinta: ¿importa el orden?, ¿se puede modificar?, ¿se repiten los
valores?, ¿necesito buscar por nombre o por posición?

## 2.1 Listas

Una **lista** (`list`) es una colección **ordenada** y **modificable** de
valores. Se escribe entre corchetes `[]`.

````python
dias_semana = ["lunes", "martes", "miércoles", "jueves", "viernes", "sábado", "domingo"]

print(dias_semana)
print(len(dias_semana))        # cuantos elementos tiene
print(dias_semana[0])          # primer elemento (indice 0)
print(dias_semana[-1])         # ultimo elemento
print(dias_semana[0:3])        # los primeros 3 (slicing, igual que en strings)
````

> En Python, **todo se cuenta desde 0**. El primer elemento de una lista de
> 7 días está en la posición `0`, y el último (el séptimo) está en la
> posición `6` (o, más fácil, en `-1`).

Las listas son **mutables**: se pueden cambiar después de creadas.

````python
unidades_vendidas = [12, 5, 40, 8, 35, 7, 15]

unidades_vendidas.append(20)        # agrega un elemento al final
print("despues de append:", unidades_vendidas)

unidades_vendidas[0] = 14           # modifica el elemento en la posicion 0
print("despues de modificar:", unidades_vendidas)

unidades_vendidas.sort()            # ordena la lista de menor a mayor, en el mismo lugar
print("ordenada:", unidades_vendidas)

print("suma total:", sum(unidades_vendidas))
print("maximo:", max(unidades_vendidas))
print("minimo:", min(unidades_vendidas))
````

## 2.2 Tuplas

Una **tupla** (`tuple`) es como una lista, pero **inmutable**: una vez
creada, no se puede modificar. Se escribe entre paréntesis `()`. Se usa para
datos que no deberían cambiar, como una coordenada o un registro fijo.

````python
coordenada_tienda = (-0.1807, -78.4678)   # (latitud, longitud) de Quito
print(coordenada_tienda)
print(coordenada_tienda[0])

try:
    coordenada_tienda[0] = 0.0
except TypeError as error:
    print("Ocurrió un TypeError:", error)
````

> Si no necesitas modificar la colección, usar una tupla es una forma de
> decirle a quien lea tu código (incluido tú, en el futuro) "esto no debería
> cambiar". No es solo una restricción técnica, también es una forma de
> comunicar intención.

## 2.3 Diccionarios

Un **diccionario** (`dict`) guarda pares **clave → valor**, no por posición
sino por nombre. Se escribe entre llaves `{}`. Es, con diferencia, la
estructura más parecida a **una fila de una tabla** — y por eso es la puerta
de entrada a Pandas.

````python
venta_lunes = {
    "dia": "lunes",
    "categoria": "Ropa",
    "unidades": 12,
    "precio_unitario": 15.0,
}

print(venta_lunes)
print(venta_lunes["categoria"])     # acceder por clave, no por posicion
print(venta_lunes.keys())           # todas las claves
print(venta_lunes.values())         # todos los valores

# Se puede agregar o modificar una clave despues de creado:
venta_lunes["ingreso"] = venta_lunes["unidades"] * venta_lunes["precio_unitario"]
print(venta_lunes)
````

Si cada día de la semana es un diccionario como el de arriba, **una semana
completa de ventas es una lista de diccionarios**: exactamente la forma en
la que normalmente pensamos una tabla, fila por fila.

````python
ventas_semana = [
    {"dia": "lunes",     "categoria": "Ropa",        "unidades": 12, "precio_unitario": 15.0},
    {"dia": "martes",    "categoria": "Electrónica", "unidades": 5,  "precio_unitario": 130.0},
    {"dia": "miércoles", "categoria": "Alimentos",   "unidades": 40, "precio_unitario": 3.0},
    {"dia": "jueves",    "categoria": "Ropa",        "unidades": 8,  "precio_unitario": 15.0},
    {"dia": "viernes",   "categoria": "Alimentos",   "unidades": 35, "precio_unitario": 3.0},
    {"dia": "sábado",    "categoria": "Electrónica", "unidades": 7,  "precio_unitario": 130.0},
    {"dia": "domingo",   "categoria": "Ropa",        "unidades": 15, "precio_unitario": 15.0},
]

print(len(ventas_semana), "dias registrados")
print(ventas_semana[0])          # la primera fila (un diccionario)
print(ventas_semana[0]["dia"])   # el valor de "dia" en la primera fila
````

Este `ventas_semana` es el ejemplo que vamos a usar durante **todo el resto
del notebook**: lo recorreremos con bucles (Sección 3), lo convertiremos en

## 2.4 Conjuntos (sets)

Un **conjunto** (`set`) guarda valores **sin orden y sin repetidos**. Sirve
para responder "¿qué valores distintos aparecen aquí?".

````python
categorias_vendidas = [venta["categoria"] for venta in ventas_semana]
print("lista completa (con repetidos):", categorias_vendidas)

categorias_unicas = set(categorias_vendidas)
print("categorias distintas:", categorias_unicas)
print("cuantas categorias distintas:", len(categorias_unicas))
````

(La línea `[venta["categoria"] for venta in ventas_semana]` es una
**comprensión de lista** — la explicamos formalmente en 3.4. Por ahora, solo
nota que recorre `ventas_semana` y se queda con la `categoria` de cada fila.)

## 2.5 Tabla comparativa: ¿cuál estructura uso?

| Estructura | ¿Ordenada? | ¿Modificable? | ¿Repite valores? | ¿Acceso por...? | Úsala cuando... |
|---|---|---|---|---|---|
| `list` | Sí | Sí | Sí | posición (`[0]`) | tienes una secuencia que puede crecer o reordenarse |
| `tuple` | Sí | No | Sí | posición (`[0]`) | el dato no debe cambiar (coordenadas, config fija) |
| `dict` | Sí (desde Python 3.7) | Sí | Valores sí, claves no | clave (`["nombre"]`) | representas **una fila** con campos con nombre |
| `set` | No | Sí | No (únicos) | no se indexa | necesitas valores únicos o comparar pertenencia |

> Esta tabla es la que vas a usar en la práctica: casi siempre vas a elegir
> entre `list` (secuencia) y `dict` (registro con nombre). `tuple` y `set`
> aparecen con menos frecuencia, pero ahora sabes reconocerlos.

# 3. Control de flujo

Hasta ahora, cada celda ejecuta sus instrucciones de arriba hacia abajo, sin
excepciones. El **control de flujo** nos deja decidir *qué* código se
ejecuta (condicionales) y *cuántas veces* (bucles).

## 3.1 Condicionales

`if` / `elif` / `else` ejecutan un bloque de código solo si una condición es
`True`. La indentación (el espacio al inicio de línea) **no es estética: es
parte de la sintaxis de Python** — define qué código pertenece a qué bloque.

````python
unidades = 5

if unidades == 0:
    estado = "sin stock"
elif unidades < 10:
    estado = "stock bajo"
else:
    estado = "stock normal"

print(f"Con {unidades} unidades, el estado es: {estado}")
````

> Python revisa las condiciones **en orden** y ejecuta solo el primer bloque
> cuya condición sea `True` (o el `else` si ninguna lo fue). Cambia
> `unidades` a `0`, a `7` y a `50` y vuelve a ejecutar la celda para
> comprobarlo.

Las condiciones se pueden combinar con los operadores lógicos de 1.2:

````python
categoria = "Electrónica"
unidades = 5

if categoria == "Electrónica" and unidades < 10:
    print("Alerta: poco stock en una categoria de alto valor")
else:
    print("Sin alerta")
````

## 3.2 Bucles `for`

Un bucle `for` repite un bloque de código **una vez por cada elemento** de
una colección (lista, tupla, string, etc.).

````python
for dia in dias_semana:
    print("Hoy es", dia)
````

Cuando necesitas también la **posición** de cada elemento, usa `enumerate`:

````python
for posicion, dia in enumerate(dias_semana):
    print(f"Día #{posicion}: {dia}")
````

`range(n)` genera los números de `0` a `n-1` — útil para repetir algo un
número exacto de veces, o para recorrer por índice.

````python
for i in range(5):
    print("Repeticion numero", i)
````

Ahora sí, el ejemplo que de verdad nos importa: recorrer `ventas_semana`
(la lista de diccionarios de 2.3) para calcular el ingreso total de la
semana.

````python
ingreso_total = 0

for venta in ventas_semana:
    ingreso_dia = venta["unidades"] * venta["precio_unitario"]
    venta["ingreso"] = ingreso_dia   # guardamos el resultado en la misma fila
    ingreso_total = ingreso_total + ingreso_dia

print(f"Ingreso total de la semana: ${ingreso_total:.2f}")
print(ventas_semana[0])   # ahora cada fila ya tiene su "ingreso"
````

## 3.3 Bucles `while`

Un bucle `while` repite un bloque **mientras una condición sea `True`**. Se
usa cuando no sabes de antemano cuántas veces vas a repetir (a diferencia de
`for`, que recorre una colección de tamaño conocido).

````python
stock = 50
unidades_por_dia = 12
dia_contador = 0

while stock > 0:
    stock = stock - unidades_por_dia
    dia_contador = dia_contador + 1

print(f"El stock se agota en {dia_contador} días (quedarían {stock} unidades, un numero negativo indica que se vendio de mas el ultimo dia)")
````

> **Cuidado con los bucles infinitos.** Un `while True:` sin una condición
> que eventualmente se vuelva `False` (o sin un `break`) nunca termina, y
> tendrás que interrumpir la celda manualmente (botón ■ / "Interrupt" del
> notebook). Es el error más frustrante de depurar para quien empieza —
> revisa siempre que la variable de la condición **cambie dentro** del
> bucle.

## 3.4 Comprensión de listas

Una **comprensión de listas** (*list comprehension*) es una forma compacta
de construir una lista nueva a partir de otra, en una sola línea. Primero,
la forma "larga" con un bucle explícito:

````python
ingresos_largo = []
for venta in ventas_semana:
    ingresos_largo.append(venta["ingreso"])

print(ingresos_largo)
````

Y ahora la misma idea como comprensión de lista:

````python
ingresos_corto = [venta["ingreso"] for venta in ventas_semana]
print(ingresos_corto)
print(ingresos_largo == ingresos_corto)   # comprobamos que son el mismo resultado
````

La leemos así: *"para cada `venta` en `ventas_semana`, quédate con
`venta['ingreso']`"*. También se puede filtrar dentro de la comprensión:

````python
ingresos_altos = [venta["ingreso"] for venta in ventas_semana if venta["ingreso"] > 150]
print(ingresos_altos)
````

> No necesitas memorizar la sintaxis de comprensión de listas hoy — basta
> con que la reconozcas cuando la veas en el Día 1 y sepas que es
> equivalente a un `for` con `.append()`.

# 4. Funciones

Ya usamos funciones todo el tiempo: `print()`, `len()`, `sum()`, `type()`.
Una **función** es un bloque de código con nombre que recibe entradas
(parámetros) y puede devolver una salida (`return`). Sirve para **no repetir
el mismo cálculo una y otra vez**.

## 4.1 Definir y llamar funciones

````python
def calcular_ingreso(unidades, precio_unitario):
    ingreso = unidades * precio_unitario
    return ingreso

# "Llamamos" a la funcion con valores concretos:
ingreso_lunes = calcular_ingreso(12, 15.0)
print(ingreso_lunes)

ingreso_martes = calcular_ingreso(5, 130.0)
print(ingreso_martes)
````

> `unidades` y `precio_unitario` dentro de la función son **parámetros**:
> nombres genéricos que reciben el valor real (el **argumento**) en cada
> llamada. `calcular_ingreso(12, 15.0)` y `calcular_ingreso(5, 130.0)`
> ejecutan el mismo código, con valores distintos cada vez — esa es toda la
> idea.

## 4.2 Parámetros por defecto y argumentos con nombre

````python
def calcular_ingreso_con_descuento(unidades, precio_unitario, descuento=0.0):
    ingreso_bruto = unidades * precio_unitario
    ingreso_neto = ingreso_bruto * (1 - descuento)
    return ingreso_neto

print(calcular_ingreso_con_descuento(12, 15.0))          # usa descuento=0.0 por defecto
print(calcular_ingreso_con_descuento(12, 15.0, 0.10))    # 10% de descuento
print(calcular_ingreso_con_descuento(unidades=12, precio_unitario=15.0, descuento=0.10))  # igual, con nombre
````

> Pasar los argumentos **con nombre** (`descuento=0.10`) no cambia el
> resultado, pero hace el código más fácil de leer, especialmente cuando una
> función tiene varios parámetros. Lo usarás mucho leyendo código de
> Pandas/Matplotlib, donde casi todo se pasa con nombre
> (`pd.read_csv(sep=';')`, `plt.hist(bins=20)`).

## 4.3 Funciones como "cajas reutilizables"

El valor real de una función aparece cuando la reusas muchas veces en vez de
copiar y pegar el mismo cálculo:

````python
for venta in ventas_semana:
    venta["ingreso_recalculado"] = calcular_ingreso(venta["unidades"], venta["precio_unitario"])

print(ventas_semana[0])
````

Si mañana cambia la fórmula del ingreso (por ejemplo, para incluir
impuestos), solo tienes que corregirla **en un lugar** (dentro de la
función) en vez de en cada cálculo repetido por el notebook.

## 4.4 Alcance de variables (scope)

Una variable creada **dentro** de una función solo existe dentro de esa
función (se dice que es "local"). Una variable creada fuera, a nivel del

````python
def demo_scope():
    variable_local = "solo existo dentro de la funcion"
    print("Dentro de la funcion:", variable_local)

demo_scope()

try:
    print(variable_local)   # esto va a fallar: variable_local no existe aqui afuera
except NameError as error:
    print("Ocurrió un NameError:", error)
````

> Este es el `NameError` que mencionamos en la tabla de errores comunes
> (Sección "¿Por qué este Día 0?"). Ahora ya sabes una de sus causas más
> comunes: usar, fuera de una función, una variable que solo existe dentro
> de ella.

# 6. NumPy a fondo

## 6.1 ¿Por qué no listas de Python para cálculo numérico?

Ya sabemos operar listas con bucles. Probemos calcular el doble de cada
unidad vendida, con una lista normal:

````python
unidades_lista = [12, 5, 40, 8, 35, 7, 15]

dobles_lista = []
for u in unidades_lista:
    dobles_lista.append(u * 2)

print(dobles_lista)
````

Funciona, pero hace falta un bucle explícito para algo tan simple como
"multiplica todo por 2". NumPy permite hacerlo **directamente sobre todo el
bloque de números a la vez** (sin escribir el `for`):

````python
unidades_array = np.array(unidades_lista)
dobles_array = unidades_array * 2

print(dobles_array)
print(type(dobles_array))
````

A esto se le llama **vectorización**: la operación se aplica a todos los
elementos "de una sola vez", sin que tú escribas el bucle (NumPy lo hace
internamente, en código optimizado). Además de ser más corto de escribir,
es mucho más rápido para datasets grandes. Comprobémoslo con un caso más
exigente:

````python
import time

n = 2_000_000
numeros_lista = list(range(n))
numeros_array = np.arange(n)

inicio = time.perf_counter()
cuadrados_lista = [x ** 2 for x in numeros_lista]
tiempo_lista = time.perf_counter() - inicio

inicio = time.perf_counter()
cuadrados_array = numeros_array ** 2
tiempo_array = time.perf_counter() - inicio

print(f"Lista de Python (comprension): {tiempo_lista:.4f} segundos")
print(f"Array de NumPy (vectorizado): {tiempo_array:.4f} segundos")
print(f"NumPy fue aproximadamente {tiempo_lista / tiempo_array:.0f} veces mas rapido")
````

Los tiempos exactos varían según la computadora, pero la relación (NumPy
decenas de veces más rápido) se mantiene siempre. La razón es de memoria,
no solo de "trucos de código":

<p align="center">
    <img src="../material_apoyo/python_lista_vs_array.png" width="700">
</p>

Una lista de Python guarda **objetos** dispersos en la memoria (y pueden ser
de tipos distintos entre sí); un `ndarray` de NumPy guarda **un bloque
contiguo** de valores, todos del mismo tipo. Por eso NumPy puede recorrerlo
y operarlo mucho más eficientemente.

> Esta es la razón por la que, en Machine Learning, casi nunca vas a ver un
> bucle `for` operando fila por fila sobre los datos: NumPy (y Pandas, que
> está construido encima de NumPy) hacen el trabajo vectorizado por ti.

## 6.2 Crear arrays

Además de `np.array()` a partir de una lista, hay funciones para crear
arrays directamente:

````python
ceros = np.zeros(5)                 # 5 ceros
unos = np.ones(5)                   # 5 unos
secuencia = np.arange(0, 20, 5)     # de 0 a 20 (sin incluir), de 5 en 5
reparticion = np.linspace(0, 1, 5)  # 5 numeros igualmente espaciados entre 0 y 1

print("zeros:    ", ceros)
print("ones:     ", unos)
print("arange:   ", secuencia)
print("linspace: ", reparticion)
````

> `np.arange(inicio, fin, paso)` es como el `range()` de Python puro, pero
> devuelve un array y acepta pasos decimales (`range()` solo acepta
> enteros). `np.linspace(inicio, fin, cantidad)` en cambio te deja elegir
> **cuántos** puntos quieres, y los reparte uniformemente.

## 6.3 `shape`, `dtype` y dimensiones

En Machine Learning trabajamos más con **matrices** (tablas de números) que
con vectores sueltos. Un array de NumPy puede tener 1, 2 o más dimensiones.

````python
vector = np.array([12, 5, 40, 8])
print("vector:", vector)
print("shape:", vector.shape)     # (4,) -> un vector de 4 elementos
print("dtype:", vector.dtype)     # el tipo de dato de TODOS los elementos

matriz = np.array([
    [12, 15.0],   # unidades, precio del lunes
    [5, 130.0],   # unidades, precio del martes
    [40, 3.0],    # unidades, precio del miercoles
])
print("matriz:\n", matriz)
print("shape:", matriz.shape)     # (3, 2) -> 3 observaciones, 2 variables
````

`shape = (3, 2)` significa **3 filas (observaciones) y 2 columnas
(variables)**. Esta notación `(filas, columnas)` es exactamente la que vas a
leer todo el Día 1 en expresiones como `X_train.shape`: es la forma
estándar de describir el tamaño de una tabla de datos.

> Nota que `matriz.dtype` va a ser `float64` aunque pusimos enteros como
> `12`: NumPy busca **un solo tipo para todo el array**, y como hay al menos
> un decimal (`15.0`), convierte todo a `float`. Esto es distinto de una
> lista de Python, que sí puede mezclar tipos libremente (como vimos en el
> diagrama de 6.1).

## 6.4 Indexado y slicing

Igual que en listas, pero con una sintaxis más cómoda para 2 dimensiones:
`array[fila, columna]`.

````python
print(matriz[0, 0])      # fila 0, columna 0: unidades del lunes
print(matriz[0, :])      # fila 0 completa (todas las columnas)
print(matriz[:, 0])      # columna 0 completa (todas las filas): todas las "unidades"
print(matriz[:, 1])      # columna 1 completa: todos los "precios"
print(matriz[0:2, :])    # las primeras 2 filas
````

> `:` solo significa "todo en esta dimensión". `matriz[:, 0]` se lee "todas
> las filas, columna 0" — es exactamente cómo se extrae **una variable (una
> columna) de toda la tabla** de una sola vez.

## 6.5 Operaciones vectorizadas y broadcasting

````python
unidades = np.array([12, 5, 40, 8, 35, 7, 15])
precios = np.array([15.0, 130.0, 3.0, 15.0, 3.0, 130.0, 15.0])

ingresos = unidades * precios   # multiplicacion elemento a elemento
print("ingresos:", ingresos)
print("ingreso total:", ingresos.sum())

# "Broadcasting": operar un array con un solo numero (escalar).
precios_con_iva = precios * 1.15   # 15% de IVA sobre cada precio
print("precios con IVA:", precios_con_iva)
````

`unidades * precios` multiplica **elemento por elemento** (la posición 0 con
la posición 0, la 1 con la 1, etc.) — no es una multiplicación de matrices
en el sentido del álgebra lineal clásica. Cuando multiplicamos un array
completo por un solo número (`precios * 1.15`), NumPy "expande" ese número
para aplicarlo a cada elemento; a eso se le llama **broadcasting**, y es lo
que nos ahorra escribir un bucle para algo tan simple como "aplica el IVA a
todos los precios".

## 6.6 Funciones estadísticas

````python
print("media:    ", ingresos.mean())
print("mediana:  ", np.median(ingresos))
print("desv.est.:", ingresos.std())
print("suma:     ", ingresos.sum())
print("minimo:   ", ingresos.min())
print("maximo:   ", ingresos.max())
````

En una matriz 2D, estas funciones aceptan un parámetro `axis` para decidir
si resumen **por fila** o **por columna**:

````python
tabla = np.array([
    [12, 180.0],   # unidades, ingreso: lunes
    [5, 650.0],    # martes
    [40, 120.0],   # miercoles
])

print("promedio de TODA la tabla:", tabla.mean())
print("promedio por columna (axis=0):", tabla.mean(axis=0))   # un promedio por variable
print("promedio por fila (axis=1):", tabla.mean(axis=1))      # un promedio por observacion
````

> `axis=0` colapsa las **filas** (un resultado por columna); `axis=1`
> colapsa las **columnas** (un resultado por fila). Si te cuesta recordarlo,
> piensa en qué eje "desaparece": con `axis=0` desaparecen los índices de
> fila, y te quedas con un valor por columna.

## 6.7 `reshape` y arrays 2D como matrices de datos

`reshape` cambia la forma de un array sin cambiar sus datos ni su cantidad
total de elementos.

````python
plano = np.arange(12)              # un vector de 12 elementos: 0,1,...,11
print("plano:", plano, "shape:", plano.shape)

como_matriz = plano.reshape(3, 4)  # 3 filas x 4 columnas = 12 elementos
print("como matriz:\n", como_matriz, "\nshape:", como_matriz.shape)

try:
    plano.reshape(3, 5)   # 3*5=15, pero plano solo tiene 12 elementos
except ValueError as error:
    print("Ocurrió un ValueError:", error)
````

> `reshape` solo funciona si el número de elementos **no cambia**: 12
> elementos se pueden acomodar en `(3, 4)`, `(4, 3)`, `(2, 6)`, `(12, 1)`...
> pero nunca en `(3, 5)` (eso serían 15 casillas para solo 12 valores). En
> el Día 1 verás esto cuando sklearn te pida convertir un vector de una sola
> variable en una matriz de una columna con `X.reshape(-1, 1)`.

# 7. Pandas a fondo

NumPy sabe operar con números puros, pero no sabe de nombres de columnas,
fechas o categorías. **Pandas** añade esa capa de significado encima de
NumPy. Es, con diferencia, la librería que más vas a usar en este curso.

## 7.1 Series: una columna con nombre

Una **Series** es un array de NumPy con una etiqueta (nombre) y un índice.

````python
precios_serie = pd.Series([15.0, 130.0, 3.0, 15.0, 3.0, 130.0, 15.0], name="precio_unitario")
print(precios_serie)
print(type(precios_serie))
print(precios_serie.mean())   # las Series heredan los mismos metodos que un array de NumPy
````

## 7.2 DataFrame: una tabla con significado

Un **DataFrame** es una colección de Series que comparten el mismo índice —
en la práctica, una tabla con filas y columnas con nombre. Se puede crear,
entre otras formas, a partir de un diccionario donde cada clave es una
columna:

````python
dias = ["lunes", "martes", "miércoles", "jueves", "viernes", "sábado", "domingo"]
categorias = ["Ropa", "Electrónica", "Alimentos", "Ropa", "Alimentos", "Electrónica", "Ropa"]
unidades = [12, 5, 40, 8, 35, 7, 15]
precios = [15.0, 130.0, 3.0, 15.0, 3.0, 130.0, 15.0]
ingresos = [u * p for u, p in zip(unidades, precios)]

df = pd.DataFrame({
    "dia": dias,
    "categoria": categorias,
    "unidades": unidades,
    "ingreso": ingresos,
})

df
````

Observa la diferencia con la matriz de NumPy de la Sección 6: ahora **cada
columna tiene nombre y tipo propio** (`dia` es texto, `unidades` es entero,
`ingreso` es decimal) en vez de que todo el bloque comparta un solo
`dtype`. El número en negrita a la izquierda (`0, 1, 2, ...`) es el
**índice**: identifica cada fila, igual que la posición en una lista — de
hecho, por defecto es simplemente `0, 1, 2, ...`, como los índices que ya

## 7.4 Selección de datos: `loc` vs `iloc`

Pandas ofrece dos formas de seleccionar datos, y confundirlas es un error
muy común al empezar.

| | `loc` | `iloc` |
|---|---|---|
| Selecciona por... | **etiqueta** (nombre de índice/columna) | **posición** (entero, como en una lista) |
| Ejemplo de fila | `df.loc[0]` (la fila *con índice* 0) | `df.iloc[0]` (la *primera* fila, sea cual sea su índice) |
| Ejemplo de columna | `df.loc[:, "unidades"]` | `df.iloc[:, 2]` |

````python
print(df.loc[0])                 # fila con etiqueta de indice 0
print(df.loc[0, "categoria"])    # esa fila, solo la columna "categoria"
print(df.loc[0:2, ["dia", "categoria"]])   # varias filas, varias columnas, por nombre
````

````python
print(df.iloc[0])                # primera fila, por posicion
print(df.iloc[0, 1])             # primera fila, segunda columna (posicion 1), por posicion
print(df.iloc[0:2, 0:2])         # primeras 2 filas, primeras 2 columnas
````

> Mientras el índice de `df` sea `0, 1, 2, ...` (el que viene por defecto),
> `loc[0]` e `iloc[0]` dan el mismo resultado — por eso la confusión es tan
> común. La diferencia se vuelve crítica cuando el índice **no** es una
> secuencia simple de enteros (por ejemplo, después de filtrar u ordenar).
> Regla práctica: usa `loc` cuando piensas en **nombres** y `iloc` cuando
> piensas en **posiciones**.

Para seleccionar **una sola columna completa**, hay un atajo más corto que
`loc`:

````python
print(df["categoria"])          # una columna -> devuelve una Series
print(df[["categoria", "ingreso"]])   # varias columnas -> devuelve un DataFrame (doble corchete)
````

## 7.5 Filtrado booleano

Para quedarte solo con las filas que cumplen una condición, se construye una
**máscara booleana** (una Series de `True`/`False`, una por fila) y se la
usa como "filtro".

````python
mascara = df["ingreso"] > 150
print(mascara)        # una Serie de True/False, una por fila

df[mascara]            # el DataFrame filtrado: solo las filas donde la mascara es True
````

````python
# Lo mismo, en una sola linea (la forma mas comun de escribirlo):
df[df["categoria"] == "Ropa"]
````

````python
# Combinando condiciones: & es "y", | es "o" (no uses 'and'/'or' con Series completas)
df[(df["categoria"] == "Electrónica") & (df["unidades"] < 10)]
````

> Con DataFrames, los operadores lógicos de Series son `&` (y), `|` (o) y
> `~` (no) — **no** `and`/`or`/`not`, que son para un solo `bool` a la vez
> (como en la Sección 1.2) y no funcionan sobre una columna entera.

> devuelve una copia.

## 7.7 Valores nulos

En datos reales, casi siempre falta algún valor. Para practicar, creamos una
**copia** de `df` con un dato faltante a propósito (no tocamos el `df`
original):

````python
df_con_nulos = df.copy()
df_con_nulos.loc[5, "unidades"] = np.nan   # "perdemos" el dato de unidades del sabado

df_con_nulos
````

````python
print(df_con_nulos.isna())          # True donde falta un dato
print(df_con_nulos.isna().sum())    # cuantos faltantes hay POR COLUMNA
````

````python
# Opcion 1: imputar (rellenar) con un valor, por ejemplo la mediana de la columna
mediana_unidades = df_con_nulos["unidades"].median()
df_relleno = df_con_nulos.fillna({"unidades": mediana_unidades})
print(df_relleno)
````

````python
# Opcion 2: eliminar las filas con faltantes
df_sin_faltantes = df_con_nulos.dropna()
print(df_sin_faltantes)
````

> Ninguna de las dos opciones es "la correcta" en abstracto — `fillna`
> conserva la fila pero inventa un valor; `dropna` es honesto sobre que no
> sabes el dato, pero pierdes toda la fila. La decisión depende del
> problema, de cuántos datos te puedes dar el lujo de perder y de si el
> faltante tiene un patrón (volverás a este tema, con más profundidad, en
> el Día 1).

# 8. Matplotlib a fondo

Una tabla te da los valores exactos; un gráfico te muestra patrones que
serían difíciles de notar solo leyendo números: tendencias, relaciones entre
variables, concentraciones, valores atípicos. **Matplotlib** es la librería
de visualización que usaremos en todo el curso.

## 8.1 Anatomía de una figura

Todo gráfico de Matplotlib tiene dos piezas: una **`Figure`** (el lienzo
completo) y uno o más **`Axes`** (cada área de dibujo individual, con sus
propios ejes, título y leyenda).

````python
fig, ax = plt.subplots(figsize=(6, 4))   # 1 Figure que contiene 1 Axes

ax.plot(df["dia"], df["ingreso"])
ax.set_title("Ingresos de la semana")
ax.set_xlabel("Día")
ax.set_ylabel("Ingreso (USD)")

plt.show()
````

<p align="center">
    <img src="../material_apoyo/matplotlib_anatomia_figura.png" width="780">
</p>

`plt.subplots()` crea la `Figure` y el (o los) `Axes` al mismo tiempo — es
la forma más flexible y la que vas a ver más seguido en el curso. Cada
método de personalización (título, etiquetas, leyenda) se llama **sobre el
`Axes`**, no sobre la `Figure`.

## 8.4 Scatter (dispersión)

Responde: *¿qué relación parece existir entre dos variables?*

````python
fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter(df["unidades"], df["ingreso"], s=80, color="#2e8b57")
ax.set_title("Unidades vendidas vs. ingreso")
ax.set_xlabel("Unidades vendidas")
ax.set_ylabel("Ingreso (USD)")
plt.show()
````

## 8.5 Boxplot

Responde: *¿cómo se distribuyen los datos por grupo, y hay valores
extremos?*

````python
datos_por_categoria = [df[df["categoria"] == cat]["ingreso"] for cat in df["categoria"].unique()]

fig, ax = plt.subplots(figsize=(6, 4))
ax.boxplot(datos_por_categoria, tick_labels=df["categoria"].unique())
ax.set_title("Ingreso por categoría")
ax.set_ylabel("Ingreso (USD)")
plt.show()
````

## 8.6 Personalización

Título, etiquetas, leyenda, color y tamaño son los cinco ajustes que vas a
repetir en prácticamente todos los gráficos del curso:

````python
fig, ax = plt.subplots(figsize=(7, 4.5))

ax.plot(df["dia"], df["ingreso"], marker="o", color="#1f6fb2", linewidth=2, label="Ingreso real")
ax.axhline(df["ingreso"].mean(), color="#e07b1a", linestyle="--", label="Ingreso promedio")

ax.set_title("Ingreso diario vs. promedio semanal", fontsize=14, fontweight="bold")
ax.set_xlabel("Día")
ax.set_ylabel("Ingreso (USD)")
ax.legend()
ax.grid(alpha=0.3)

plt.show()
````

> `label="..."` en `plot()` es lo que hace que `ax.legend()` sepa qué texto
> mostrar por cada serie. Si no le pones `label`, esa línea simplemente no
> aparece en la leyenda, aunque sí se dibuje.

## 8.7 Varios gráficos a la vez

`plt.subplots(filas, columnas)` crea una **grilla** de `Axes` dentro de una
sola `Figure` — así se comparan varias vistas sin saturar un solo gráfico.

````python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))

axes[0].plot(df["dia"], df["ingreso"], marker="o", color="#1f6fb2")
axes[0].set_title("Ingreso por día")
axes[0].set_xlabel("Día")
axes[0].set_ylabel("Ingreso (USD)")

axes[1].scatter(df["unidades"], df["ingreso"], color="#2e8b57", s=80)
axes[1].set_title("Unidades vs. ingreso")
axes[1].set_xlabel("Unidades")
axes[1].set_ylabel("Ingreso (USD)")

fig.tight_layout()
plt.show()
````

> `axes` ahora es un **array de NumPy de objetos `Axes`** (sí, el mismo
> `ndarray` de la Sección 6): `axes[0]` es el primero, `axes[1]` el
> segundo. `fig.tight_layout()` ajusta automáticamente los espacios para
> que los títulos y etiquetas no se encimen entre subgráficos.

# 9. Integración: de Python puro a un DataFrame graficado

Cerramos uniendo las cuatro piezas con el mismo ejemplo, de principio a fin,
en una sola celda — exactamente el recorrido que vas a repetir, con datos
reales, en el Día 1.

````python
# 1) Python puro: una lista de diccionarios (una fila = un dia de ventas).
ventas_integracion = [
    {"dia": "lunes",     "categoria": "Ropa",        "unidades": 12, "precio_unitario": 15.0},
    {"dia": "martes",    "categoria": "Electrónica", "unidades": 5,  "precio_unitario": 130.0},
    {"dia": "miércoles", "categoria": "Alimentos",   "unidades": 40, "precio_unitario": 3.0},
    {"dia": "jueves",    "categoria": "Ropa",        "unidades": 8,  "precio_unitario": 15.0},
    {"dia": "viernes",   "categoria": "Alimentos",   "unidades": 35, "precio_unitario": 3.0},
    {"dia": "sábado",    "categoria": "Electrónica", "unidades": 7,  "precio_unitario": 130.0},
    {"dia": "domingo",   "categoria": "Ropa",        "unidades": 15, "precio_unitario": 15.0},
]

# 2) NumPy: extraemos columnas numericas como arrays y calculamos el ingreso vectorizado.
unidades_np = np.array([fila["unidades"] for fila in ventas_integracion])
precios_np = np.array([fila["precio_unitario"] for fila in ventas_integracion])
ingresos_np = unidades_np * precios_np
print("Ingreso total (NumPy):", ingresos_np.sum())

# 3) Pandas: armamos un DataFrame y lo exploramos/resumimos.
df_final = pd.DataFrame(ventas_integracion)
df_final["ingreso"] = ingresos_np
print(df_final.describe())
print(df_final.groupby("categoria")["ingreso"].sum())

# 4) Matplotlib: lo graficamos.
fig, ax = plt.subplots(figsize=(7, 4.5))
ax.bar(df_final["dia"], df_final["ingreso"], color="#1f6fb2")
ax.set_title("Ingreso diario de la tienda (flujo completo)")
ax.set_xlabel("Día")
ax.set_ylabel("Ingreso (USD)")
plt.show()
````

Repasa el camino que acabas de recorrer:

```text
lista de diccionarios (Python puro)
        │
        ▼
arrays de NumPy (unidades, precios) -> operacion vectorizada -> ingresos
        │
        ▼
DataFrame de Pandas (tabla con nombre, describe(), groupby())
        │
        ▼
grafico de Matplotlib (una pregunta visual respondida)
```
