# 1. Train, Validation y Test: qué decisión toma cada conjunto

```text
TRAIN       -> aprende parámetros
VALIDATION  -> decide hiperparámetros, variables y transformaciones
TEST        -> comprueba al final
```

Un **parámetro** es aprendido por el modelo. Ejemplo: coeficientes de regresión o reglas de un árbol.

Un **hiperparámetro** lo elegimos nosotros o un procedimiento de búsqueda. Ejemplo: `max_depth`.

## 1.1 Overfitting y underfitting

| Train | Validation | Lectura |
|---|---|---|
| excelente | mala | overfitting |
| mala | mala similar | underfitting o features insuficientes |
| buena | buena similar | generalización razonable |

No basta mirar una única métrica de un único conjunto.

````python
# Simulación conceptual: no es un entrenamiento real, sino una tabla para aprender a diagnosticar.
diagnostico = pd.DataFrame({
    'modelo':['A','B','C'],
    'R2_train':[0.96,0.30,0.82],
    'R2_validation':[0.38,0.27,0.79],
    'lectura':['Overfitting','Underfitting','Generalización razonable']
})
diagnostico
````

## 1.2 Data leakage

Leakage ocurre cuando el entrenamiento recibe información que **no estaría disponible en el momento real de predecir**.

Ejemplos:

- usar una variable calculada con el resultado final;
- normalizar todo el dataset antes del split y dejar que Validation influya en la media/desviación;
- incluir registros duplicados de la misma persona en Train y Test;
- para forecasting, usar información del futuro.

Un resultado demasiado bueno no es prueba de leakage, pero sí una razón para investigarlo.

# 2. Validación cruzada

Un único Validation puede ser afortunado o desafortunado. K-Fold Cross Validation rota qué observaciones actúan como validación.

Con `k=5`:

1. se divide Train en 5 bloques;
2. se entrena 5 veces;
3. cada bloque actúa una vez como validación;
4. se promedian las métricas.

**Test sigue fuera del proceso.**

````python
# Ejemplo con un dataset de regresión pequeño para ver la mecánica.
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
from sklearn.metrics import make_scorer, mean_absolute_error

X_reg, y_reg = make_regression(n_samples=250, n_features=5, noise=15, random_state=42)
modelo = LinearRegression()

resultado_cv = cross_validate(
    modelo,
    X_reg,
    y_reg,
    cv=5,
    scoring=['neg_mean_absolute_error','r2'],
    return_train_score=True
)

pd.DataFrame(resultado_cv)
````

````python
# Las pérdidas vienen negativas por convención de scoring de sklearn: mayor = mejor.
mae_val = -resultado_cv['test_neg_mean_absolute_error']
r2_val = resultado_cv['test_r2']

print('MAE por fold:', np.round(mae_val,2))
print('MAE promedio:', round(mae_val.mean(),2))
print('R² promedio :', round(r2_val.mean(),3))
````

## 2.1 Cross-validation no reemplaza Test

Un error frecuente es pensar que, si usamos 5-Fold CV, ya no necesitamos Test. En realidad cumplen papeles distintos:

- **Cross-validation** ayuda a estimar estabilidad y seleccionar decisiones dentro del proceso de desarrollo.
- **Test** permanece apartado para comprobar el diseño final.

Si hacemos 200 pruebas de hiperparámetros usando el mismo CV, también podemos terminar adaptándonos a ese procedimiento de validación. Por eso conviene mantener un conjunto final independiente.

````python
# Resumen de variabilidad entre folds.
resumen_cv = pd.DataFrame({
    'MAE_validation': mae_val,
    'R2_validation': r2_val
})

print(resumen_cv.round(3))
print()
print('Desviación del MAE entre folds:', round(resumen_cv['MAE_validation'].std(), 3))
````

### ¿Qué nos dice la dispersión entre folds?

Si la métrica cambia mucho entre folds, puede significar:

- dataset pequeño;
- grupos heterogéneos;
- outliers;
- modelo inestable;
- split que deja subpoblaciones muy diferentes.

No debemos mirar únicamente el promedio.

# 3. Cambiamos de regresión a clasificación

En regresión el target es continuo:

```text

```

En clasificación el target es una etiqueta:


La pregunta ya no es “¿cuántos puntos me equivoqué?”, sino “¿qué tipos de casos estoy confundiendo?”.

### La variable `duration` y el concepto de leakage temporal/operacional

En Bank Marketing, `duration` representa la duración de la llamada. Según el momento real en que queramos predecir, esta variable puede ser problemática:

- si predecimos **antes de llamar**, todavía no conocemos la duración;
- si predecimos **después de la llamada**, sí podría estar disponible.

La disponibilidad de una feature depende del instante de decisión, no solo de si existe en el CSV.

# 4. Preparación transparente para el árbol

Para entender el modelo, hoy usaremos `pd.get_dummies` en lugar de ocultar todo dentro de un Pipeline.

Pasos:

1. retirar columnas auxiliares como `tipo_dato`;
2. separar X e y;
3. convertir categorías a 0/1;
4. separar Train/Validation/Test.

````python
datos = df_bank.copy()

# Eliminamos una marca del respaldo si existe.
datos = datos.drop(columns=['tipo_dato'], errors='ignore')

# Convertimos target a 0/1.
y = datos['y'].astype(str).str.lower().map({'yes':1,'no':0})
X = datos.drop(columns=['y'])

# One-hot para columnas de texto.
X_enc = pd.get_dummies(X, drop_first=True, dtype=int)

print('X original:', X.shape)
print('X codificado:', X_enc.shape)
X_enc.head()
````

````python
# Separación 70/15/15 aproximada con estratificación del target.
X_train_val, X_test, y_train_val, y_test = train_test_split(
    X_enc, y,
    test_size=0.15,
    stratify=y,
    random_state=42
)

X_train, X_val, y_train, y_val = train_test_split(
    X_train_val, y_train_val,
    test_size=0.1765,
    stratify=y_train_val,
    random_state=42
)

print(X_train.shape, X_val.shape, X_test.shape)
print('Proporción positiva Train:', y_train.mean().round(3))
print('Proporción positiva Val  :', y_val.mean().round(3))
````

## 4.1 Un baseline de clasificación

Antes del árbol, construimos un clasificador trivial: predecir siempre la clase mayoritaria.

Si el dataset está desbalanceado, este baseline puede tener Accuracy sorprendentemente alta. Esa es justamente la razón por la que necesitamos otras métricas.

````python
# Clase mayoritaria aprendida SOLO desde Train.
clase_mayoritaria = int(y_train.mode().iloc[0])
pred_baseline = np.full(len(y_val), clase_mayoritaria)

print('Clase mayoritaria:', clase_mayoritaria)
print('Accuracy baseline:', round(accuracy_score(y_val, pred_baseline), 3))
print('Recall baseline  :', round(recall_score(y_val, pred_baseline, zero_division=0), 3))
print('F1 baseline      :', round(f1_score(y_val, pred_baseline, zero_division=0), 3))
````

# 5. Decision Tree: aprender reglas

Un árbol hace preguntas como:

```text
duration <= 250 ?
    sí -> ...
    no -> ...
```

En cada nodo prueba cortes candidatos y busca una división que deje grupos más “puros” respecto a la clase.

En clasificación, criterios comunes son **Gini** y **Entropy**.

## 5.1 Gini con un ejemplo mínimo

Si un nodo tiene 50% sí y 50% no:

\[
Gini = 1 - (0.5^2 + 0.5^2) = 0.5
\]

Si todos son de la misma clase, Gini = 0: nodo puro.

````python
def gini(proporciones):
    proporciones = np.asarray(proporciones)
    return 1 - np.sum(proporciones**2)

print('50/50 ->', gini([0.5,0.5]))
print('90/10 ->', gini([0.9,0.1]))
print('100/0 ->', gini([1.0,0.0]))
````

### Entropía como criterio alternativo

Otra medida de impureza es la entropía:

\[
H = -\sum p_i \log_2(p_i)
\]

- nodo puro -> entropía 0;
- mezcla 50/50 de dos clases -> entropía máxima de 1.

Gini y Entropy suelen producir árboles parecidos, aunque no siempre idénticos.

````python
def entropia(proporciones):
    p = np.asarray(proporciones, dtype=float)
    p = p[p > 0]  # evitamos log(0)
    return -np.sum(p * np.log2(p))

print('50/50 ->', entropia([0.5,0.5]))
print('90/10 ->', entropia([0.9,0.1]))
print('100/0 ->', entropia([1.0,0.0]))
````

## 5.2 Primer árbol pequeño

Empezamos con `max_depth=3` para poder dibujarlo y entenderlo.

````python
arbol = DecisionTreeClassifier(
    max_depth=3,
    min_samples_leaf=25,
    random_state=42
)

arbol.fit(X_train, y_train)
pred_val_tree = arbol.predict(X_val)

print('Accuracy Val:', round(accuracy_score(y_val, pred_val_tree),3))
print('F1 Val      :', round(f1_score(y_val, pred_val_tree),3))
````

````python
# Dibujamos el árbol. Con muchas features, los nombres pueden ser largos.
plt.figure(figsize=(18,8))
plot_tree(
    arbol,
    feature_names=X_train.columns,
    class_names=['No','Sí'],
    filled=True,
    max_depth=3,
    fontsize=7
)
plt.show()
````

## 5.3 Qué ocurre cuando aumenta `max_depth`

Un árbol muy profundo puede crear reglas extremadamente específicas y memorizar Train.

Vamos a variar profundidad y comparar Train vs Validation.

````python
resultados = []

for profundidad in range(1, 16):
    modelo = DecisionTreeClassifier(
        max_depth=profundidad,
        random_state=42
    )
    modelo.fit(X_train, y_train)

    resultados.append({
        'max_depth':profundidad,
        'acc_train':accuracy_score(y_train, modelo.predict(X_train)),
        'acc_val':accuracy_score(y_val, modelo.predict(X_val)),
        'f1_val':f1_score(y_val, modelo.predict(X_val))
    })

resultados = pd.DataFrame(resultados)
resultados
````

````python
plt.figure(figsize=(8,4))
plt.plot(resultados['max_depth'], resultados['acc_train'], marker='o', label='Train')
plt.plot(resultados['max_depth'], resultados['acc_val'], marker='o', label='Validation')
plt.xlabel('max_depth')
plt.ylabel('Accuracy')
plt.title('Profundidad vs generalización')
plt.legend()
plt.show()
````

### Hiperparámetros principales de un árbol

| Hiperparámetro | Controla |
|---|---|
| `max_depth` | profundidad máxima |
| `min_samples_split` | mínimo para dividir un nodo |
| `min_samples_leaf` | mínimo permitido por hoja |
| `criterion` | cómo se evalúa la calidad de una división |
| `ccp_alpha` | poda por costo-complejidad |

Si Train sube y Validation cae, aumentar complejidad ya no está ayudando a generalizar.

## 5.4 `min_samples_leaf`: una forma intuitiva de regularizar

Una hoja con 1 o 2 observaciones puede capturar detalles demasiado específicos. Aumentar `min_samples_leaf` obliga a que cada regla final tenga más evidencia.

````python
filas_leaf=[]
for leaf in [1,2,5,10,25,50,100]:
    m=DecisionTreeClassifier(max_depth=None,min_samples_leaf=leaf,random_state=42)
    m.fit(X_train,y_train)
    filas_leaf.append({
        'min_samples_leaf':leaf,
        'acc_train':accuracy_score(y_train,m.predict(X_train)),
        'f1_val':f1_score(y_val,m.predict(X_val),zero_division=0)
    })

pd.DataFrame(filas_leaf).round(3)
````

## 5.5 Poda por costo-complejidad (`ccp_alpha`)

En lugar de limitar únicamente la profundidad, podemos permitir que el árbol crezca y después penalizar ramas cuya mejora sea pequeña.

````python
# Tomamos algunos alphas de la ruta de poda para no generar una tabla enorme.
path = DecisionTreeClassifier(random_state=42).cost_complexity_pruning_path(X_train,y_train)
alphas = np.unique(np.quantile(path.ccp_alphas, [0, .25, .5, .75, .9]))

filas=[]
for alpha in alphas:
    m=DecisionTreeClassifier(random_state=42,ccp_alpha=float(alpha))
    m.fit(X_train,y_train)
    filas.append({
        'ccp_alpha':float(alpha),
        'n_hojas':m.get_n_leaves(),
        'profundidad':m.get_depth(),
        'F1_val':f1_score(y_val,m.predict(X_val),zero_division=0)
    })

pd.DataFrame(filas).round(4)
````

# 6. Random Forest: muchos árboles diferentes

Random Forest crea muchos árboles con variaciones de filas y variables y luego combina sus decisiones.

Idea intuitiva:

```text
Árbol 1 -> Sí
Árbol 2 -> No
Árbol 3 -> Sí
Árbol 4 -> Sí
Árbol 5 -> No
----------------
Mayoría -> Sí
```

La diversidad entre árboles reduce la varianza de un único árbol.

````python
forest = RandomForestClassifier(
    n_estimators=200,     # cantidad de árboles
    max_depth=8,          # complejidad de cada árbol
    min_samples_leaf=5,
    random_state=42,
    n_jobs=-1
)

forest.fit(X_train, y_train)
pred_val_rf = forest.predict(X_val)

print('Accuracy:', round(accuracy_score(y_val, pred_val_rf),3))
print('Precision:', round(precision_score(y_val, pred_val_rf, zero_division=0),3))
print('Recall:', round(recall_score(y_val, pred_val_rf, zero_division=0),3))
print('F1:', round(f1_score(y_val, pred_val_rf, zero_division=0),3))
````

## 6.0 Bootstrap y `max_features`

Dos fuentes de diversidad son importantes:

- **bootstrap:** cada árbol recibe una muestra con reemplazo de las filas;
- **max_features:** en cada división no siempre puede mirar todas las variables.

Si todos los árboles fueran idénticos, promediarlos aportaría poco.

````python
# Mini ejemplo de bootstrap sobre índices 0..9.
rng_demo=np.random.default_rng(42)
for arbol_id in range(3):
    muestra=rng_demo.choice(np.arange(10),size=10,replace=True)
    print(f'Árbol {arbol_id+1}:',muestra)
````

## 6.1 ¿Más árboles siempre mejora?

Aumentar `n_estimators` suele estabilizar la predicción, pero llega un punto de rendimientos decrecientes. No tiene el mismo efecto que aumentar indefinidamente la profundidad.

````python
comparacion_n = []

for n in [10, 30, 60, 100, 200, 400]:
    m = RandomForestClassifier(
        n_estimators=n,
        max_depth=8,
        min_samples_leaf=5,
        random_state=42,
        n_jobs=-1
    )
    m.fit(X_train,y_train)
    p = m.predict(X_val)
    comparacion_n.append({'n_estimators':n,'F1_val':f1_score(y_val,p,zero_division=0)})

pd.DataFrame(comparacion_n)
````

## 6.2 Importancia de variables: útil, pero no causal

`feature_importances_` muestra qué variables ayudaron a reducir impureza dentro de los árboles. Es una medida predictiva interna, no una estimación causal.

````python
importancia = pd.Series(
    forest.feature_importances_,
    index=X_train.columns
).sort_values(ascending=False)

importancia.head(15)
````

## 6.3 Efecto de `max_features`

Este hiperparámetro controla cuántas variables puede considerar cada división. Menos features puede aumentar diversidad, aunque demasiada restricción también puede debilitar cada árbol.

````python
filas=[]
for mf in ['sqrt','log2',0.5,1.0]:
    m=RandomForestClassifier(
        n_estimators=200,max_depth=8,min_samples_leaf=5,
        max_features=mf,random_state=42,n_jobs=-1
    )
    m.fit(X_train,y_train)
    p=m.predict(X_val)
    filas.append({'max_features':str(mf),'F1_val':f1_score(y_val,p,zero_division=0)})

pd.DataFrame(filas).round(3)
````

## 6.4 `class_weight` cuando importa detectar la minoría

`class_weight='balanced'` aumenta el costo relativo de equivocarse en clases poco frecuentes. No es una solución mágica, pero puede cambiar el equilibrio Precision/Recall.

````python
rf_balanceado=RandomForestClassifier(
    n_estimators=250,max_depth=8,min_samples_leaf=5,
    class_weight='balanced',random_state=42,n_jobs=-1
)
rf_balanceado.fit(X_train,y_train)
p_bal=rf_balanceado.predict(X_val)

def metricas_clasif_local(y_real,pred):
    return pd.Series({
        'Accuracy':accuracy_score(y_real,pred),
        'Precision':precision_score(y_real,pred,zero_division=0),
        'Recall':recall_score(y_real,pred,zero_division=0),
        'F1':f1_score(y_real,pred,zero_division=0)
    })

pd.DataFrame({
    'RF_normal':metricas_clasif_local(y_val,pred_val_rf),
    'RF_balanced':metricas_clasif_local(y_val,p_bal)
}).T.round(3)
````

# 7. Comparación de modelos del día

Mañana profundizaremos en métricas de clasificación. Por ahora guardamos una tabla comparable.

````python
def resumen_clasificacion(y_real, pred):
    return pd.Series({
        'Accuracy':accuracy_score(y_real,pred),
        'Precision':precision_score(y_real,pred,zero_division=0),
        'Recall':recall_score(y_real,pred,zero_division=0),
        'F1':f1_score(y_real,pred,zero_division=0)
    })

pd.DataFrame({
    'DecisionTree':resumen_clasificacion(y_val,pred_val_tree),
    'RandomForest':resumen_clasificacion(y_val,pred_val_rf)
}).T.round(3)
````

````python
# Matriz de confusión del Random Forest.
ConfusionMatrixDisplay.from_predictions(
    y_val,
    pred_val_rf,
    display_labels=['No','Sí']
)
plt.title('Random Forest - Validation')
plt.show()
````
