````python
# Dataset sintetico propio de este notebook: ventas diarias de una tienda,
# un poco mas grande que el del Dia 0 para que los ejemplos de sklearn
# tengan sentido (sklearn necesita varias filas para "aprender" algo).
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

%matplotlib inline

rng = np.random.default_rng(42)   # semilla fija: mas adelante explicamos por que

n = 20
categorias = rng.choice(["Ropa", "Electrónica", "Alimentos"], size=n, p=[0.4, 0.3, 0.3])
precio_base = {"Ropa": 15.0, "Electrónica": 130.0, "Alimentos": 3.0}
unidades = rng.integers(3, 45, size=n)
precio_unitario = np.array([precio_base[c] for c in categorias])
ruido = rng.normal(0, 5, size=n)
ingreso = unidades * precio_unitario + ruido
es_fin_de_semana = rng.choice([True, False], size=n, p=[0.3, 0.7])

df = pd.DataFrame({
    "categoria": categorias,
    "unidades": unidades,
    "precio_unitario": precio_unitario,
    "ingreso": ingreso.round(2),
    "es_fin_de_semana": es_fin_de_semana,
})

df.head()
````

Nada nuevo hasta aquí: `pd.DataFrame`, `.head()`, operaciones vectorizadas —
todo del Día 0. Lo único nuevo es `np.random.default_rng(42)`, que genera
números "aleatorios" de forma reproducible. Volvemos sobre esto en la
Sección 1.3.

# 1. El patrón `fit` / `predict` de scikit-learn

## 1.1 Todo modelo de scikit-learn se usa igual

Scikit-learn tiene decenas de modelos distintos (regresión lineal, árboles,
KNN, SVM, K-Means...). La buena noticia: **todos se usan con la misma receta
de tres pasos**, sin importar cuál sea:

```text
1) crear el modelo   ->  modelo = AlgunModelo(hiperparametros)
2) entrenarlo        ->  modelo.fit(X, y)
3) usarlo            ->  modelo.predict(X_nuevo)
```

Una vez que reconoces esta receta, reconoces el 80% del código de cualquier
notebook del curso, sin importar qué modelo use.

````python
from sklearn.linear_model import LinearRegression

# Paso 1: crear el modelo (todavia no sabe nada, son solo "instrucciones vacias")
modelo = LinearRegression()

# Paso 2: entrenarlo con datos conocidos (X = entradas, y = lo que queremos predecir)
X = df[["unidades"]]   # OJO: doble corchete. Lo explicamos en 1.2.
y = df["ingreso"]
modelo.fit(X, y)

# Paso 3: usarlo para predecir sobre datos nuevos (aqui, los mismos X de ejemplo)
predicciones = modelo.predict(X)
print(predicciones[:5])
````

> Nota que **ni el modelo ni `fit` imprimen nada por sí solos** — `fit`
> modifica el objeto `modelo` "por dentro" (le calcula una pendiente y un
> intercepto, igual que en el Día 1) pero no devuelve un resultado para
> imprimir. `predict`, en cambio, sí devuelve algo: un array de NumPy con
> una predicción por cada fila de `X`.

Cualquier modelo de sklearn que veas en el curso —`DecisionTreeRegressor`,
`RandomForestClassifier`, `KNeighborsClassifier`, `KMeans`, `PCA`— se maneja
con esta misma receta. Cambia el nombre del modelo y, a veces, si el paso 3
se llama `predict` o `transform` — pero la estructura `crear → fit → usar`
es siempre la misma.

## 1.2 Por qué `X` siempre debe ser una tabla (2D), nunca un vector suelto

En el bloque anterior escribimos `df[["unidades"]]` con **doble corchete**,
no `df["unidades"]`. No es un descuido: scikit-learn exige que `X` tenga
forma de tabla, aunque sea de una sola columna.

````python
print("con doble corchete (DataFrame, 2D):", type(df[["unidades"]]), df[["unidades"]].shape)
print("con un solo corchete (Series, 1D):  ", type(df["unidades"]), df["unidades"].shape)

try:
    modelo_fallido = LinearRegression()
    modelo_fallido.fit(df["unidades"], df["ingreso"])   # 1D: esto falla
except ValueError as error:
    print("\nOcurrió un ValueError:", error)
````

> `(20,)` es la forma de un vector de 20 elementos (1D); `(20, 1)` es la
> forma de una tabla de 20 filas y 1 columna (2D) — la misma distinción de
> `.shape` que viste en el Día 0 (Sección 6.3), aplicada ahora a por qué
> sklearn rechaza un vector suelto. `y`, en cambio, **sí** puede ser 1D: solo
> `X` tiene esta exigencia, porque `X` puede tener muchas columnas y sklearn
> necesita esa estructura de tabla siempre, incluso con una sola variable.

Si ya tienes un array de NumPy en vez de un DataFrame, la forma equivalente
es `.reshape(-1, 1)` (ya lo viste en el Día 0, Sección 6.7):

````python
unidades_array = df["unidades"].to_numpy()     # 1D: shape (20,)
unidades_2d = unidades_array.reshape(-1, 1)    # 2D: shape (20, 1)
print(unidades_array.shape, "->", unidades_2d.shape)
````

> `-1` en `reshape(-1, 1)` significa "calcula tú ese número" — le estamos
> diciendo a NumPy "quiero 1 columna, acomoda las filas que hagan falta".
> Es exactamente lo mismo que harás con `X.reshape(-1, 1)` si alguna vez
> trabajas con un solo array de NumPy en vez de un DataFrame.

## 1.3 Reproducibilidad: semillas (`random_state`)

Varias piezas de scikit-learn usan números "al azar" (separar Train/Test,
inicializar un árbol, etc.). El problema: si el azar es realmente azaroso,
**cada vez que ejecutes el notebook obtendrías un resultado distinto**, y
nadie podría reproducir lo que hiciste. La solución es fijar una **semilla**.

````python
# Sin semilla fija, dos "sorteos" dan resultados distintos:
print("sorteo 1 (sin semilla):", np.random.default_rng().integers(0, 100, 3))
print("sorteo 2 (sin semilla):", np.random.default_rng().integers(0, 100, 3))

# Con la MISMA semilla, el sorteo es identico, siempre:
print("sorteo con semilla 42:  ", np.random.default_rng(42).integers(0, 100, 3))
print("sorteo con semilla 42:  ", np.random.default_rng(42).integers(0, 100, 3))
````

> `random_state=42` que verás en `train_test_split`, en modelos como
> `RandomForestRegressor` o en `KMeans` hace exactamente esto: fija la
> semilla para que **tu resultado sea reproducible** — si vuelves a correr
> la celda (o alguien más corre tu notebook), obtienen exactamente el mismo
> split, el mismo árbol, los mismos clusters. El número en sí (`42`, `0`,
> `7`...) no importa — lo único que importa es que sea **el mismo número
> cada vez** que quieras reproducir el resultado.

# 2. `train_test_split` a fondo

Ya viste el concepto de Train/Validation/Test en el Día 1. Aquí nos
enfocamos solo en la **función de Python** que hace la separación.

````python
from sklearn.model_selection import train_test_split

X = df[["unidades", "precio_unitario"]]
y = df["ingreso"]

resultado = train_test_split(X, y, test_size=0.25, random_state=42)
print(type(resultado), len(resultado))
````

`train_test_split` devuelve **cuatro tablas a la vez**, en un único objeto
(una tupla de 4 elementos). Por eso casi siempre se ve escrito así, usando
desempaquetado de tupla (lo mismo que `for dia, venta in ...` en el Día 0,
pero para una asignación en vez de un bucle):

````python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

print("X_train:", X_train.shape)
print("X_test: ", X_test.shape)
print("y_train:", y_train.shape)
print("y_test: ", y_test.shape)
````

> El **orden** en el que escribes los nombres a la izquierda debe coincidir
> con el orden en el que `train_test_split` los devuelve:
> `X_train, X_test, y_train, y_test` — siempre en ese orden (primero ambos
> `X`, luego ambos `y`). Si los escribes en otro orden, Python no se queja
> (no hay error), pero terminarías entrenando con datos de prueba sin
> saberlo — un bug silencioso, no un error que veas en pantalla.

### Parámetros que vas a ver siempre

| Parámetro | Qué controla |
|---|---|
| `test_size=0.25` | qué fracción de las filas va a Test (aquí, 25%) |
| `random_state=42` | la semilla (Sección 1.3): mismo split cada vez que corras la celda |
| `stratify=columna` | (lo verás en el Día 3) mantiene las mismas proporciones de categorías en Train y Test — útil cuando `y` es una categoría desbalanceada |

# 3. Diccionarios y tuplas como "configuración" de un experimento

Ya conoces listas, tuplas y diccionarios del Día 0. En los notebooks del
curso aparecen combinados de formas que todavía no viste, casi siempre para
**comparar varios modelos a la vez** sin repetir código.

## 3.1 Lista de tuplas + desempaquetado en un `for`

Una forma muy común de guardar "varias configuraciones" es una lista donde
cada elemento es una tupla `(nombre, objeto)`.

````python
modelos_a_comparar = [
    ("Regresión con unidades", LinearRegression()),
    ("Regresión con precio", LinearRegression()),
]

X_unidades = df[["unidades"]]
X_precio = df[["precio_unitario"]]
entradas = [X_unidades, X_precio]

for (nombre, modelo), X_actual in zip(modelos_a_comparar, entradas):
    modelo.fit(X_actual, df["ingreso"])
    print(f"{nombre}: coeficiente = {modelo.coef_[0]:.3f}")
````

> `for (nombre, modelo), X_actual in zip(...)` hace dos cosas a la vez:
> **desempaqueta** cada tupla `(nombre, modelo)` en dos variables, y usa
> `zip()` (del Día 0, recíbelo como "cremallera": une dos listas elemento a
> elemento) para recorrer `modelos_a_comparar` y `entradas` juntas, en
> paralelo. Si te resulta denso, puedes escribir el equivalente más largo:
> un `for i in range(len(modelos_a_comparar))` e indexar con `[i]` — llega
> al mismo resultado, solo que con más líneas.

## 3.2 Diccionario de configuraciones

Para guardar más de dos cosas por configuración (no solo nombre y modelo),
es común usar un diccionario de diccionarios, o un diccionario de tuplas:

````python
configuraciones = {
    "simple": {"columnas": ["unidades"], "color": "#1f6fb2"},
    "completa": {"columnas": ["unidades", "precio_unitario"], "color": "#e07b1a"},
}

for clave, config in configuraciones.items():
    modelo = LinearRegression()
    modelo.fit(df[config["columnas"]], df["ingreso"])
    print(f"Configuración '{clave}' usa columnas {config['columnas']} (color {config['color']})")
````

## 3.3 El doble asterisco `**`: desempaquetar un diccionario dentro de otro

Esta es, con diferencia, la sintaxis menos intuitiva que vas a encontrar en
el curso — y aparece todo el tiempo para armar tablas de resultados. Vamos
paso a paso.

````python
base = {"modelo": "Regresión simple"}
metricas = {"MAE": 1.2, "R2": 0.45}

# Forma larga, sin ** (la que ya conoces):
fila_larga = {}
fila_larga.update(base)
fila_larga.update(metricas)
print("forma larga:", fila_larga)

# Forma corta, con ** (la que veras en los notebooks del curso):
fila_corta = {**base, **metricas}
print("forma corta:", fila_corta)

print(fila_larga == fila_corta)
````

> `{**base, **metricas}` se lee: *"crea un diccionario nuevo, copiando
> adentro todas las claves de `base`, y luego todas las claves de
> `metricas`"*. Es exactamente lo mismo que las dos líneas de `.update()` de
> arriba, pero en una sola expresión. Si una clave se repitiera en ambos
> diccionarios, gana la que se escribió **después** — igual que si llamaras
> `.update()` en ese mismo orden.

Ahora el patrón que vas a reconocer en los notebooks de Día 1, 2 y 5, para
armar una tabla comparando modelos:

````python
def calcular_metricas(y_real, y_pred):
    error_abs = np.abs(y_real - y_pred)
    return {"MAE": error_abs.mean(), "MAE_max": error_abs.max()}

resultados = []
for nombre, columnas in [("Solo unidades", ["unidades"]), ("Unidades y precio", ["unidades", "precio_unitario"])]:
    modelo = LinearRegression()
    modelo.fit(df[columnas], df["ingreso"])
    pred = modelo.predict(df[columnas])
    fila = {"modelo": nombre, **calcular_metricas(df["ingreso"], pred)}
    resultados.append(fila)

pd.DataFrame(resultados)
````

> Léelo de afuera hacia adentro: `resultados` es una lista de diccionarios
> (ya lo viste en el Día 0 — es justo la forma `ventas_semana`), y cada
> diccionario se arma con `{"modelo": nombre, **calcular_metricas(...)}`:
> *"una clave `modelo` con el nombre, más todas las claves que devuelva
> `calcular_metricas`"*. Al final, `pd.DataFrame(resultados)` convierte esa
> lista de diccionarios en una tabla — el mismo truco que ya conocías,
> aplicado a comparar modelos.

*Reto 3*: agrega una tercera configuración a la lista (`["precio_unitario"]`
solo) y vuelve a correr la celda. Después, modifica `calcular_metricas` para

# 4. `Pipeline`: encadenar pasos sin errores de fuga de datos

## 4.1 El problema que resuelve

Antes de entrenar un modelo, casi siempre hay que **transformar** los datos:
escalar variables numéricas, convertir categorías con One-Hot, imputar
faltantes. El riesgo es mezclar el orden: si transformas **antes** de
separar Train/Test, el Test "contamina" el aprendizaje de la transformación
— esto es la *fuga de datos* (data leakage) que ya se menciona en los
notebooks del curso. Un `Pipeline` empaqueta "transformar + modelar" como
**un solo objeto**, para que ese orden nunca se rompa por accidente.

## 4.2 Sintaxis: lista de tuplas (nombre, paso)

````python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

pipeline = Pipeline([
    ("escalar", StandardScaler()),      # paso 1: estandariza las columnas numericas
    ("modelo", LinearRegression()),     # paso 2: el modelo final
])

X = df[["unidades", "precio_unitario"]]
y = df["ingreso"]

pipeline.fit(X, y)                      # fit entrena TODOS los pasos, en orden
predicciones = pipeline.predict(X)      # predict pasa los datos por TODOS los pasos
print(predicciones[:5])
````

> Fíjate que `pipeline.fit(X, y)` y `pipeline.predict(X)` son **exactamente
> la misma receta** `fit`/`predict` de la Sección 1 — un `Pipeline` también
> es, en sí mismo, un estimador de scikit-learn, solo que por dentro corre
> varios pasos en cadena en vez de uno solo. Esa es la idea más importante
> de esta sección: `Pipeline` no es un concepto nuevo de uso, es el mismo
> patrón aplicado a varios pasos a la vez.

## 4.3 `named_steps`: acceder a un paso interno

Una vez entrenado el `Pipeline`, puedes "entrar" a cualquiera de sus pasos
por su nombre, para inspeccionarlo:

````python
print(pipeline.named_steps["modelo"].coef_)        # los coeficientes del paso "modelo"
print(pipeline.named_steps["escalar"].mean_)        # las medias que aprendio el escalador
````

## 4.4 La convención `paso__parametro` (doble guion bajo)

Si quieres cambiar un hiperparámetro de un paso específico **desde fuera**
del `Pipeline` (por ejemplo, para probar varias opciones, como en la
Sección 5), scikit-learn usa el nombre del paso, dos guiones bajos, y el
nombre del parámetro:

````python
from sklearn.linear_model import Ridge

pipeline_ridge = Pipeline([
    ("escalar", StandardScaler()),
    ("modelo", Ridge()),
])

# "modelo__alpha" = "el parametro alpha DEL PASO llamado modelo"
pipeline_ridge.set_params(modelo__alpha=10.0)
pipeline_ridge.fit(X, y)
print(pipeline_ridge.named_steps["modelo"].alpha)
````

> `modelo__alpha` no es un nombre de variable con un error de tipeo — los
> dos guiones bajos son **sintaxis real** de scikit-learn, no estética.
> Dentro de un `Pipeline` con un paso llamado `"modelo"`, cualquier
> parámetro de ese paso se direcciona como `modelo__nombre_del_parametro`.
> Vas a ver esto todo el tiempo en los días 1.1, 3, 4, 6 y 7 — y en la
> Sección 5 de este notebook, dentro de un `param_grid`.

# 5. Buscar el mejor modelo automáticamente: `GridSearchCV`

## 5.1 Qué automatiza

Hasta ahora, para comparar configuraciones usamos un `for` escrito a mano
(Sección 3). `GridSearchCV` hace lo mismo, pero automáticamente: prueba
**todas las combinaciones** de los hiperparámetros que le indiques, y para
cada una repite la validación varias veces (K-Fold, que ya viste
conceptualmente en el Día 2).

## 5.2 El diccionario `param_grid`

````python
from sklearn.model_selection import GridSearchCV

pipeline_busqueda = Pipeline([
    ("escalar", StandardScaler()),
    ("modelo", Ridge()),
])

# Cada clave usa la convención paso__parametro de la Seccion 4.4.
param_grid = {
    "modelo__alpha": [0.1, 1.0, 10.0, 50.0],
}

busqueda = GridSearchCV(
    pipeline_busqueda,
    param_grid=param_grid,
    cv=4,                              # 4-Fold, como en el Dia 2
    scoring="neg_mean_absolute_error",
)
busqueda.fit(X, y)
print("Mejor alpha encontrado:", busqueda.best_params_)
````

> Lee `param_grid` como una receta: *"para el paso `modelo` dentro del
> pipeline, prueba cada uno de estos 4 valores de `alpha`"*. `GridSearchCV`
> entrena y valida el pipeline completo una vez por cada valor (aquí, 4
> veces), y se queda con el que obtuvo mejor resultado en la validación.

## 5.3 El signo negativo: por qué `neg_mean_absolute_error`

Scikit-learn tiene una regla interna: **"mayor score es siempre mejor"**.
Pero el MAE funciona al revés (menor error es mejor). La solución de
scikit-learn es, simplemente, invertir el signo del MAE antes de
compararlo, para que la regla "mayor es mejor" se siga cumpliendo:

````python
print("mejor score encontrado (negativo):", busqueda.best_score_)
print("ese mismo valor, como MAE real:   ", -busqueda.best_score_)
````

> Por eso vas a ver código como `-cv_resultado['test_MAE']` o
> `scoring='neg_mean_absolute_error'` en los días 2 y 5: no es un error de
> signo en el notebook, es la convención de scikit-learn para que todas sus
> herramientas de búsqueda (`GridSearchCV`, `cross_val_score`,
> `RandomizedSearchCV`) puedan comparar cualquier métrica con la misma regla
> ("mayor es mejor"), sin importar si esa métrica originalmente premiaba
> valores altos (como R²) o bajos (como MAE).

## 5.4 Atributos de un objeto ya entrenado

Después de `.fit()`, `busqueda` queda "cargado" de información adicional,
accesible como atributos (sin paréntesis, porque no son funciones a
llamar, son datos ya calculados):

````python
print("best_params_:", busqueda.best_params_)
print("best_score_: ", busqueda.best_score_)

# cv_results_ es un diccionario: se puede convertir directo en DataFrame
resultados_busqueda = pd.DataFrame(busqueda.cv_results_)[["param_modelo__alpha", "mean_test_score"]]
resultados_busqueda
````

> El guion bajo al final (`best_params_`, `best_score_`, `cv_results_`) es
> una convención de scikit-learn: significa *"este atributo solo existe
> después de llamar a `.fit()`"*. Si intentas leer `busqueda.best_params_`
> **antes** de entrenar, obtienes un error — es la forma que tiene
> scikit-learn de avisarte "todavía no he aprendido nada".

# 6. `assert` y leer un `Traceback` más largo

El Día 0 ya te presentó tres errores comunes (`NameError`, `SyntaxError`,
`TypeError`) con `try`/`except`. En los notebooks del curso (sobre todo
Días 4, 6 y 7) vas a ver, además, la palabra `assert` usada como "chequeo de
seguridad" dentro del código, sin ningún `try`/`except` alrededor.

## 6.1 Qué es un `assert`

`assert condicion, "mensaje"` comprueba que `condicion` sea `True`. Si lo
es, no pasa nada (el código sigue). Si es `False`, Python **detiene la
ejecución** y muestra el mensaje — a propósito, porque algo que el autor del
código daba por hecho resultó ser falso.

````python
total_filas = len(df)
assert total_filas > 0, "El DataFrame no debería estar vacío"
print("El assert pasó sin problema: hay", total_filas, "filas")
````

````python
try:
    assert total_filas > 1000, "Se esperaban más de 1000 filas, pero el dataset es más chico"
except AssertionError as error:
    print("Ocurrió un AssertionError (a propósito, para que veas el mensaje):", error)
````

> En los notebooks del curso **no** vas a ver el `try/except` alrededor del
> `assert` (lo pusimos aquí solo para que el notebook no se detenga). En la
> práctica, un `assert` que falla es una señal intencional de "algo en tus
> datos o tu lógica no es lo que esperabas, revisa antes de seguir" — por
> ejemplo, confirmar que Train y Test no comparten ninguna fila. No es un
> error de programación que tengas que arreglar en el código; es una alarma
> sobre los **datos**.

## 6.2 Leer un `Traceback` más largo

Cuando un error ocurre **dentro** de una función que llamaste, Python te
muestra **toda la cadena de llamadas**, no solo la línea final. Veamos un
ejemplo con dos funciones, una que llama a la otra:

````python
def dividir(a, b):
    return a / b

def calcular_promedio_por_unidad(ingreso_total, unidades_totales):
    return dividir(ingreso_total, unidades_totales)

try:
    calcular_promedio_por_unidad(100, 0)   # division por cero, a proposito
except ZeroDivisionError as error:
    print("Ocurrió un ZeroDivisionError:", error)
````

> Si quitaras el `try/except` de la celda anterior, el `Traceback` mostraría
> **tres niveles**: la línea donde llamaste a
> `calcular_promedio_por_unidad(100, 0)`, luego la línea dentro de esa
> función donde llama a `dividir(...)`, y finalmente la línea dentro de
> `dividir` donde ocurre la división. Regla práctica: **lee el Traceback de
> abajo hacia arriba** — la última línea te dice *qué* pasó (el tipo de
> error y el mensaje), y las líneas de arriba te dicen *dónde*, en qué
> cadena de funciones, ocurrió.

## 7.1 `ax` se vuelve un array cuando pides más de un gráfico

````python
fig, ax = plt.subplots(1, 2, figsize=(10, 4))   # 1 fila, 2 columnas de graficos

print(type(ax), ax.shape)   # ax ahora es un array de NumPy de 2 "Axes"

ax[0].hist(df["ingreso"], bins=6, color="#1f6fb2")
ax[0].set_title("Histograma de ingresos")

ax[1].scatter(df["unidades"], df["ingreso"], color="#e07b1a")
ax[1].set_title("Unidades vs. ingreso")

fig.tight_layout()
plt.show()
````

> Esto es, literalmente, el mismo `ndarray` de NumPy del Día 0: `ax[0]` es
> el primer gráfico, `ax[1]` el segundo. Si pides una grilla de 2×2 en vez
> de 1×2 (`plt.subplots(2, 2)`), `ax` se vuelve un array **2D**, y accedes a
> cada panel con `ax[0, 0]`, `ax[0, 1]`, `ax[1, 0]`, `ax[1, 1]` — la misma
> notación `[fila, columna]` de un array 2D que ya conoces.

## 7.2 Un vistazo rápido a un gráfico en 3D

Esto lo vas a ver una sola vez, en el Día 1, para visualizar una regresión
con dos variables de entrada (un plano, en vez de una recta). No hace falta
que lo domines — solo que reconozcas las tres piezas nuevas.

````python
from mpl_toolkits.mplot3d import Axes3D   # habilita projection='3d'

fig = plt.figure(figsize=(7, 5))
ax3d = fig.add_subplot(111, projection="3d")   # "111" = 1 fila, 1 columna, grafico 1

ax3d.scatter(df["unidades"], df["precio_unitario"], df["ingreso"], color="#2e8b57")
ax3d.set_xlabel("Unidades")
ax3d.set_ylabel("Precio unitario")
ax3d.set_zlabel("Ingreso")
ax3d.set_title("Un scatter en 3D: ahora hay un tercer eje (Z)")

plt.show()
````

> Tres piezas nuevas, nada más: `projection="3d"` (le pide a Matplotlib un
> espacio de 3 ejes en vez de 2), `"111"` (una forma compacta y antigua de
> escribir "1 fila, 1 columna, gráfico número 1" — equivalente a
> `plt.subplots(1, 1)`), y un tercer eje `Z` además de `X` e `Y`. Si en el
> Día 1 ves además `plot_surface(...)` y `np.meshgrid(...)`, son para
> dibujar un **plano** (la superficie de una regresión con 2 variables) en
> vez de puntos sueltos — no necesitas reproducir ese código a mano, solo
> reconocer que sigue siendo "un gráfico, con un eje extra".

# 8. Integración: un mini-experimento de punta a punta

Cerramos uniendo todas las piezas de este notebook en un solo flujo —
exactamente el tipo de celda "densa" que vas a encontrar a partir del Día 3.

````python
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge

# 1) Separar Train/Test (Seccion 2), con semilla fija (Seccion 1.3).
X = df[["unidades", "precio_unitario"]]
y = df["ingreso"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

# 2) Pipeline de 2 pasos (Seccion 4).
pipeline = Pipeline([
    ("escalar", StandardScaler()),
    ("modelo", Ridge()),
])

# 3) Busqueda de hiperparametros con la convencion paso__parametro (Seccion 5).
param_grid = {"modelo__alpha": [0.1, 1.0, 10.0]}
busqueda = GridSearchCV(pipeline, param_grid=param_grid, cv=4, scoring="neg_mean_absolute_error")
busqueda.fit(X_train, y_train)

# 4) Tabla de resultados armada con **dict-unpacking (Seccion 3.3).
pred_test = busqueda.predict(X_test)
mae_test = np.abs(y_test - pred_test).mean()
resumen = [{"etapa": "Entrenamiento (CV)", "MAE": -busqueda.best_score_},
           {"etapa": "Test", "MAE": mae_test}]
tabla_resumen = pd.DataFrame(resumen)
print("Mejor alpha:", busqueda.best_params_)
tabla_resumen
````

````python
# 5) Grafico de 2 paneles (Seccion 7): valores reales vs. predichos, y error por fila.
fig, ax = plt.subplots(1, 2, figsize=(11, 4))

ax[0].scatter(y_test, pred_test, color="#1f6fb2")
ax[0].plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], "--", color="gray")
ax[0].set_title("Real vs. predicho (Test)")
ax[0].set_xlabel("Ingreso real")
ax[0].set_ylabel("Ingreso predicho")

residuos = y_test.to_numpy() - pred_test
ax[1].scatter(range(len(residuos)), residuos, color="#e07b1a")
ax[1].axhline(0, color="gray", linestyle="--")
ax[1].set_title("Residuos en Test")
ax[1].set_xlabel("Observación")
ax[1].set_ylabel("Error (real - predicho)")

fig.tight_layout()
plt.show()
````

Repasa qué reconociste en esas dos celdas: `train_test_split` con
`random_state` (Sección 2), un `Pipeline` de dos pasos (Sección 4),
`GridSearchCV` con `param_grid` usando la convención `paso__parametro`
(Sección 5), una lista de diccionarios para armar la tabla de resultados
(Sección 3), y una figura con dos paneles (Sección 7). Si hace un momento
cada una de esas piezas por separado ya tenía sentido, esta celda "densa"
debería leerse ahora como una combinación de cosas conocidas, no como código
nuevo.
