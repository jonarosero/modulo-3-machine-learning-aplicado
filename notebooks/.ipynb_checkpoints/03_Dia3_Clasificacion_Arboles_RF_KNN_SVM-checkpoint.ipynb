## Antes de empezar: ¿qué es un problema de clasificación?

En aprendizaje supervisado tenemos ejemplos del pasado. Cada ejemplo contiene datos de entrada —que llamamos X— y una respuesta conocida —que llamamos y—. Entrenamos un modelo con esos ejemplos para que, al recibir una fila nueva, estime una respuesta.

<p align="center">
    <img src="../material_apoyo/classification.png" width="350">
</p>

La primera pregunta es: ¿qué tipo de respuesta queremos obtener?

- En regresión, la respuesta es una cantidad numérica. Por ejemplo: «¿cuánto crecerá el PIB?». Una predicción podría ser 2,4 %. Nos importa qué tan lejos está ese número del valor real: predecir 2,4 % cuando ocurrió 2,6 % es un error menor que predecir 8 %.
- En clasificación, la respuesta es una categoría. Por ejemplo: «¿el cliente contratará un depósito?». Las respuestas posibles son yes y no. El modelo debe asignar una de esas clases a cada contacto nuevo.

Imagina que conocemos de una persona su edad, ocupación y antecedentes de contacto. Si preguntamos «¿cuánto dinero depositará?», buscamos una cantidad y planteamos una regresión. Si preguntamos «¿contratará el depósito: sí o no?», buscamos una categoría y planteamos una clasificación. Los datos de entrada podrían parecerse; lo que define el tipo de problema es la respuesta que queremos predecir.

En este notebook haremos clasificación binaria, porque hay dos clases: no y yes. Llamaremos clase positiva a yes, ya que queremos estudiar especialmente cuántos posibles suscriptores detectamos. «Positiva» no significa «buena» ni implica que el modelo esté seguro: es el nombre que damos a la clase de interés. Otros problemas pueden tener más de dos clases; por ejemplo, clasificar una solicitud como «riesgo bajo», «medio» o «alto».

Un clasificador aprende relaciones entre los datos de entrada y las respuestas históricas. No conoce por sí solo el costo de equivocarse. Decir yes cuando la respuesta real era no puede causar un contacto innecesario; decir no cuando era yes puede hacer perder una oportunidad. Por eso no bastará con contar aciertos: más adelante examinaremos qué tipos de errores comete.

Finalmente, debemos definir cuándo se usará la predicción. Aquí queremos decidir antes de iniciar una llamada a quién contactar. Podemos usar datos disponibles en ese momento, pero no duration, que indica cuánto duró la llamada y solo se conoce después. Incluirla sería darle al modelo una pista del futuro que no tendrá cuando lo utilicemos de verdad.

### Una clase no es lo mismo que una probabilidad

Supongamos que queremos decidir si contactamos a una persona para ofrecerle un depósito. Al pasar sus datos al modelo, predict() puede devolver yes. Eso significa: «según la regla de decisión del modelo, esta persona pertenece a la clase de quienes contratarían». No significa que vaya a contratar con certeza. Del mismo modo, una respuesta no no significa que sea imposible que contrate.

¿Cómo llega el modelo a esa respuesta? Algunos clasificadores calculan primero un valor intermedio o score: una señal que ayuda a situar a la persona de un lado u otro de la frontera de decisión. Un score mayor puede indicar más evidencia a favor de yes, pero no debemos leerlo automáticamente como una probabilidad. Por ejemplo, un score de 2 no quiere decir «2 %» ni «dos veces más probable».

Algunos modelos también pueden entregar una probabilidad estimada, como 0,70. Se interpretaría como una estimación del 70 % bajo ese modelo, no como una garantía de que esa persona contratará. Además, esas probabilidades pueden necesitar una comprobación de calibración: entre muchas personas a quienes se les asignó aproximadamente 70 %, ¿contrató realmente cerca del 70 %?

Finalmente, una regla de decisión convierte la señal del modelo en yes o no. Si hacemos más fácil asignar yes, normalmente detectaremos más posibles suscriptores, pero también contactaremos a más personas que no contratarán. Si somos más exigentes para asignar yes, haremos menos contactos innecesarios, pero podríamos dejar pasar oportunidades. En este notebook, SVM (SVC) usa su frontera de decisión predeterminada; su salida 0 o 1 es una clase, no una probabilidad de 0 % o 100 %.

### Error de clasificación y decisiones económicas

Imaginemos dos personas. A la primera, el modelo le asigna yes, la llamamos y no contrata: hicimos un contacto que no produjo una suscripción. Esto es un falso positivo. A la segunda, el modelo le asigna no, no la llamamos, pero sí habría contratado: dejamos pasar una posible oportunidad. Esto es un falso negativo.

No siempre conviene reducir ambos errores de la misma manera. Si llamar cuesta poco y perder una suscripción potencial cuesta mucho, podríamos aceptar más contactos innecesarios con tal de encontrar más personas que sí contratarían. Más adelante veremos que Recall ayuda a responder: «de quienes realmente eran yes, ¿a cuántos encontramos?». En cambio, si cada llamada requiere mucho tiempo o dinero, nos interesará que las personas señaladas como yes tengan más posibilidades de contratar. Precision ayuda a responder: «de quienes marcamos como yes, ¿cuántos realmente lo eran?».

Por eso, el modelo con más aciertos totales no es necesariamente el más útil. Primero debemos definir qué decisión se tomará, cuánto cuesta cada tipo de error y cuántos contactos podemos realizar. Después miraremos la matriz de confusión y las métricas para saber qué modelo se ajusta mejor a esa necesidad. Los datos históricos nos ayudan a comparar opciones, pero el algoritmo no decide por nosotros cuánto vale una llamada o una oportunidad perdida.

### ¿Qué es el desbalance de clases y por qué importa?
Hay desbalance de clases cuando una respuesta aparece mucho más que la otra. Aquí casi 89 de cada 100 contactos terminaron en no, mientras que solo 11 de cada 100 terminaron en yes. no es la clase mayoritaria y yes la minoritaria. Nos interesa especialmente detectar yes, por lo que la llamaremos clase positiva; «positiva» es solo una etiqueta de análisis, no un juicio sobre las personas.

Consideremos una regla muy simple: responder siempre no, sin mirar edad, ocupación ni ninguna otra columna. Esta regla se llama modelo de referencia o baseline: sirve para saber qué se consigue sin un algoritmo sofisticado. Acertaría aproximadamente el 88,73 % de los contactos, porque esa es la proporción de no. Sin embargo, no encontraría ni uno solo de los 4 640 yes. Su porcentaje de aciertos sería alto, pero no ayudaría a identificar posibles suscriptores. Por eso el desbalance cambia la forma en que debemos evaluar los modelos.

### No todos los «aciertos» responden a la misma pregunta

Accuracy significa proporción de aciertos totales: de todos los contactos, ¿en cuántos coincidió la predicción con la respuesta real? Es fácil de entender, pero en estos datos puede quedar dominada por la abundancia de no.

Para evaluar la clase yes, distinguiremos dos errores:
- Falso positivo: el modelo predijo yes, pero el resultado histórico fue no. En el ejemplo de la campaña, representa una alerta que no terminó en contratación.
- Falso negativo: el modelo predijo no, pero el resultado histórico fue yes. Representa un contacto positivo que el modelo no identificó como tal.
Con esa distinción podemos formular preguntas más útiles:
- Precision: de todos los contactos que el modelo marcó como yes, ¿qué proporción era realmente yes? Ayuda a valorar cuántas alertas positivas fueron acertadas.
- Recall: de todos los contactos que realmente eran yes, ¿qué proporción encontró el modelo? Ayuda a valorar cuántos positivos dejó escapar.
- F1: reúne Precision y Recall en una sola cifra y penaliza que una de las dos sea muy baja. Es útil para resumir ese equilibrio, pero no decide por sí sola qué tipo de error cuesta más.

Por ejemplo, un modelo podría marcar a muy pocas personas como yes y acertar casi todas: tendría buena Precision, pero podría dejar escapar muchos suscriptores y tener bajo Recall. Otro podría encontrar a más suscriptores, pero generar más alertas equivocadas. Veremos los números concretos en la matriz de confusión, una tabla que cruza la respuesta real con la predicción.

### Una variable que no podemos utilizar: duration
duration indica cuánto duró la llamada. Podría ayudar mucho a distinguir yes de no en los registros históricos, pero nuestro problema consiste en decidir antes de llamar. En ese momento todavía no sabemos cuánto durará la conversación.

Usar duration sería una fuga de información (data leakage): dar al modelo, durante el entrenamiento y la evaluación, una pista que no estará disponible cuando deba tomar la decisión real. Podríamos obtener resultados muy buenos en el notebook que luego no podríamos reproducir al usar el modelo. Por eso excluimos esa columna antes de entrenar.

## Preparar una comparación común y sin fuga

Antes de entrenar, fijamos **una sola muestra estratificada** de hasta 12 000 contactos para que cuatro algoritmos sean comparables en un computador de clase. Dividimos Train (70 %), Validation (15 %) y Test (15 %), conservando aproximadamente la frecuencia de `yes`. El preprocesamiento se aprende únicamente con Train dentro de cada Pipeline; Test queda cerrado hasta la decisión final.


### ¿Por qué conservar tres subconjuntos?

**Train** aprende parámetros del preprocesamiento y del modelo. **Validation** sirve para comparar árboles, bosque, `k`, kernel y `C`; como consultamos Validation para elegir, deja de ser completamente independiente. **Test** simula datos no usados durante el desarrollo y se reserva hasta que el criterio esté fijado.

La muestra estratificada de 12 000 filas mantiene la proporción `yes/no` y reduce el tiempo de entrenamiento en clase; el CSV original contiene 41 188 contactos. Los índices concretos se fijan con una semilla. Por ello los resultados describen este muestreo y no son necesariamente idénticos a entrenar con todas las filas.

```python
# Una muestra única: evita enfrentar métricas obtenidas con tamaños de entrenamiento distintos.
datos = bank.groupby("y", group_keys=False).sample(frac=min(12000 / len(bank), 1), random_state=42)
y = datos["y"].map({"no": 0, "yes": 1})  # 1 será la clase positiva en todas las métricas.
X = datos.drop(columns=["y", "duration"])  # duration aún no se conoce antes de la llamada.
X_train, X_rest, y_train, y_rest = train_test_split(X, y, test_size=.30, random_state=42, stratify=y)
X_val, X_test, y_val, y_test = train_test_split(X_rest, y_rest, test_size=.50, random_state=42, stratify=y_rest)
num = X.select_dtypes(include="number").columns.tolist()
cat = X.select_dtypes(exclude="number").columns.tolist()
# Distancias y márgenes necesitan escalado; categorías se transforman sin inventarles orden.
pre = ColumnTransformer([
    ("num", Pipeline([("impute", SimpleImputer(strategy="median")), ("scale", StandardScaler())]), num),
    ("cat", Pipeline([("impute", SimpleImputer(strategy="most_frequent")),
                       ("onehot", OneHotEncoder(handle_unknown="ignore"))]), cat)
])
# Los árboles usan el mismo X/particiones, pero no necesitan escalar números para cortar por umbrales.
pre_arbol = ColumnTransformer([
    ("num", SimpleImputer(strategy="median"), num),
    ("cat", Pipeline([("impute", SimpleImputer(strategy="most_frequent")),
                       ("onehot", OneHotEncoder(handle_unknown="ignore"))]), cat)
])
print("Train / Validation / Test:", X_train.shape, X_val.shape, X_test.shape)
print("Proporción yes en Train/Val/Test:", [round(float(s.mean()), 3) for s in (y_train, y_val, y_test)])
```

### Qué hace el preprocesamiento, paso a paso

1. Separamos los datos crudos antes de aprender transformaciones.
2. En columnas numéricas, la mediana rellena faltantes y `StandardScaler` resta la media y divide por la desviación estándar de Train.
3. Para árboles, la imputación numérica es la misma pero no se escala: cambiar unidades no cambia el orden de un umbral.
4. En columnas categóricas, la categoría más frecuente rellena faltantes y One-Hot crea indicadores binarios; no asigna un orden artificial como `admin=1`, `technician=2`.
5. El Pipeline ajusta todo en Train y luego aplica exactamente esas reglas a Validation/Test.

Así, el escalador no «mira» el futuro para calcular media/desviación ni aparecen categorías de Test durante el ajuste. Este patrón se llama evitar fuga de preprocesamiento.

### Un límite del escalado y del One-Hot

Estandarizar ayuda a que las variables numéricas no dominen una distancia solo por sus unidades. No convierte las categorías en continuas: con One-Hot, dos categorías diferentes quedan a una distancia fija según la codificación. KNN y SVM pueden cambiar bastante si cambian las variables o la codificación. Por eso documentamos las decisiones y contrastamos si tienen sentido, en lugar de interpretar el resultado como una verdad natural del dataset.

### ¿Por qué el Pipeline recibe datos originales?

`X_train` aún contiene columnas numéricas y categóricas en su forma legible. `ColumnTransformer` usa las listas `num` y `cat` para enviar cada columna por la ruta correcta. Al ejecutar `.fit(X_train, y_train)`, el transformador aprende estadísticos y categorías; el estimador final aprende una frontera. Al pedir `.predict(X_val)`, solo se aplican transformaciones ya aprendidas.

Mantener el proceso en una única estructura disminuye el riesgo de que un paso manual se aplique distinto en cada partición.

## Baseline antes del primer árbol

Un baseline es una referencia deliberadamente simple. Aquí predice siempre la clase más frecuente, `no`. Si su Accuracy parece alta, compara el Recall de `yes` antes de celebrar: un modelo que no detecta un solo suscriptor no resuelve la pregunta. Las métricas se calcularán formalmente después de obtener las predicciones del árbol; por ahora observa las cuentas de la clase positiva.


### Clase mayoritaria y clase positiva

En la codificación `no → 0`, `yes → 1`, el promedio de `y` es la proporción de `yes`. `DummyClassifier(strategy='most_frequent')` ignora todos los predictores y siempre devuelve la clase más común. Es intencionalmente sencillo: permite descubrir si un algoritmo complejo mejora una estrategia trivial.

**Predicción no es probabilidad:** `predict()` devuelve una clase; algunos modelos también devuelven scores o probabilidades. Si la clase positiva es minoritaria, un clasificador puede acertar muchos `no` y no detectar ningún `yes`. Por eso reportamos métricas por clase y comparamos con el baseline.

```python
# La referencia mayoritaria ignora X: la usamos para no elogiar una Accuracy engañosa.
baseline = DummyClassifier(strategy="most_frequent").fit(X_train, y_train)
pred_base = baseline.predict(X_val)
print("Frecuencia real en Validation:", y_val.value_counts().sort_index().to_dict())
print("Predicciones del baseline:", pd.Series(pred_base).value_counts().sort_index().to_dict())
```

### Lee el baseline antes de ajustar algoritmos

La clase `no` ocupa cerca del 89 % de esta muestra. Como el baseline siempre responde `no`, acertará muchos negativos pero detectará **cero** `yes`. Esta observación bastará por ahora; tras construir una matriz de confusión definiremos una función común para medir baseline y cuatro clasificadores con Accuracy, Precision, Recall y F1.


### Pregunta de comprensión

Si el objetivo fuera encontrar posibles suscriptores, ¿aceptarías un modelo que acierta casi nueve de diez contactos pero nunca detecta un `yes`? Explica qué información falta para decidir si contactar de más es preferible a omitir una oportunidad.


## Árbol de clasificación: preguntas que separan clases

En regresión, una hoja entregaba un promedio numérico. Ahora cada fila tiene una **clase**. Un árbol pregunta, por ejemplo, si `age` supera un umbral o si una categoría codificada por One-Hot está presente. Las observaciones siguen ramas hasta llegar a una hoja; la clase más frecuente en esa hoja será la predicción habitual. Un árbol profundo puede memorizar Train: `max_depth` y `min_samples_leaf` controlan su complejidad.

**Impureza Gini:** si una hoja tiene proporciones $p_0$ de `no` y $p_1$ de `yes`, $G=1-p_0^2-p_1^2$. Vale 0 si todas las filas son de una sola clase y 0,5 si hay mitad y mitad. Al proponer una pregunta, el árbol busca que la impureza **ponderada por tamaño** de las nuevas ramas sea menor. Gini no es Accuracy: guía el aprendizaje del árbol; las métricas posteriores evalúan las predicciones.

**Ejemplo manual:** en diez contactos con cinco `no` y cinco `yes`, $G=0,5$. Si una pregunta produce una rama de cinco `no` (Gini 0) y otra de cinco `yes` (Gini 0), su impureza ponderada es 0: esa separación sería perfecta en esas diez filas. En datos reales las ramas suelen seguir mezcladas. Una mayoría `no` en la raíz también explica por qué una hoja puede favorecer a esa clase minorando el Recall de `yes`.


```python
# Cálculo transparente: primero mezcla 50/50, luego dos ramas puras.
def gini(conteos):
    proporciones = np.asarray(conteos) / sum(conteos)
    return 1 - np.sum(proporciones ** 2)

print("Gini raíz (5 no, 5 yes):", gini([5, 5]))
print("Gini ponderado de ramas puras:", (5 * gini([5, 0]) + 5 * gini([0, 5])) / 10)
```

### Ajustar y leer el primer árbol

Usamos las mismas filas de Train y Validation que emplearán los demás modelos. El Pipeline imputa valores y hace One-Hot solo con Train. Un árbol no necesita que `age` se estandarice para preguntar por un umbral; por eso usa `pre_arbol` en lugar del preprocesador escalado de KNN/SVM. Limitamos profundidad y tamaño de hoja para evitar una regla exclusiva por contacto. **Aún no vemos Test.**


```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

arbol = Pipeline([
    ("prep", pre_arbol),
    ("model", DecisionTreeClassifier(max_depth=5, min_samples_leaf=30, random_state=42)),
])
arbol.fit(X_train, y_train)  # Aprende imputación, categorías y divisiones en Train.
pred_arbol_val = arbol.predict(X_val)  # Validation no reajusta ningún paso.
print("Profundidad aprendida:", arbol.named_steps["model"].get_depth())
print("Hojas aprendidas:", arbol.named_steps["model"].get_n_leaves())
print("Predicciones yes en Validation:", int(pred_arbol_val.sum()))
```

### Antes de mirar un porcentaje

Las hojas y la profundidad describen complejidad, no utilidad. El conteo de `yes` predichos tampoco indica cuántos eran correctos. Necesitamos cruzar las etiquetas **reales** con las **predichas**: de ahí sale la matriz de confusión. Si el árbol produce pocos `yes`, pregunta si el desbalance de Train y el umbral de clase pueden explicar parte del comportamiento.


## Métricas desde la matriz de confusión

Tomamos `yes` como clase positiva. Al cruzar clase real con predicción aparecen cuatro casos: **VP** (sí real y sí predicho), **VN** (no real y no predicho), **FP** (predijo sí, pero era no) y **FN** (predijo no, pero era sí).

$$Accuracy=\frac{VP+VN}{VP+VN+FP+FN},\quad Precision=\frac{VP}{VP+FP}$$

$$Recall=\frac{VP}{VP+FN},\quad F1=2\frac{Precision\cdot Recall}{Precision+Recall}$$

Accuracy resume aciertos globales; Precision pregunta qué tan confiables son las alertas positivas; Recall pregunta cuántos positivos reales encontramos; F1 resume el equilibrio entre Precision y Recall. Ninguna métrica determina por sí sola cuánto vale un error en dinero, tiempo o riesgo.

```python
# Ejemplo pequeño para calcular las métricas a mano
VP, VN, FP, FN = 18, 72, 8, 12
accuracy_demo = (VP + VN) / (VP + VN + FP + FN)
precision_demo = VP / (VP + FP)
recall_demo = VP / (VP + FN)
f1_demo = 2 * precision_demo * recall_demo / (precision_demo + recall_demo)
print({"Accuracy": round(accuracy_demo, 3),
       "Precision": round(precision_demo, 3),
       "Recall": round(recall_demo, 3),
       "F1": round(f1_demo, 3)})
```

**Comprueba el ejemplo:** había 30 positivos reales (`VP+FN`), pero el clasificador encontró 18; por eso Recall es 18/30. Marcó 26 observaciones como positivas (`VP+FP`), de las cuales 18 eran correctas; por eso Precision es 18/26. Si una campaña busca aprovechar oportunidades, un FN puede costar una oportunidad perdida; si cada contacto tiene un costo, también importan los FP. La prioridad depende del uso.

```python
# Orden [0, 1]: filas reales (no, yes), columnas predichas (no, yes).
matriz_arbol = confusion_matrix(y_val, pred_arbol_val, labels=[0, 1])
vn, fp, fn, vp = matriz_arbol.ravel()
print(pd.DataFrame(matriz_arbol, index=["real no", "real yes"],
                   columns=["pred no", "pred yes"]))
print("VN, FP, FN, VP:", vn, fp, fn, vp)
ConfusionMatrixDisplay.from_predictions(y_val, pred_arbol_val, display_labels=["no", "yes"])
plt.title("Árbol: errores en Validation")
plt.show()
```

### Las mismas métricas para toda la comparación

Accuracy = `(VP+VN)/total` mide aciertos globales. Precision de `yes` = `VP/(VP+FP)` mide la calidad de las alertas; Recall = `VP/(VP+FN)` mide cuántos positivos reales encontramos; F1 equilibra las dos últimas. La función siguiente aplica estas cuatro definiciones a **la misma clase positiva y el mismo Validation**. `zero_division=0` permite describir el baseline que no predice ningún `yes`, sin fingir que tuvo una Precision informativa.


```python
def resumen(y_real, y_pred, nombre):
    # y_real siempre contiene clases reales; y_pred, las predicciones del modelo.
    return {"modelo": nombre,
            "accuracy": accuracy_score(y_real, y_pred),
            "precision_yes": precision_score(y_real, y_pred, zero_division=0),
            "recall_yes": recall_score(y_real, y_pred, zero_division=0),
            "f1_yes": f1_score(y_real, y_pred, zero_division=0)}

resultados = [resumen(y_val, pred_base, "Baseline"),
              resumen(y_val, pred_arbol_val, "Árbol")]
modelos = {"Árbol": arbol}
pd.DataFrame(resultados).set_index("modelo").round(3)
```

## Random Forest: varios árboles para una misma clase

En lugar de depender de un solo árbol, el bosque entrena muchos árboles sobre muestras con reemplazo (*bootstrap*) y subconjuntos de variables. Combina sus votos para predecir una clase. Puede estabilizar reglas, pero **no corrige automáticamente el desbalance**: revisa Recall y falsos negativos, no solo Accuracy. `n_estimators` indica cuántos árboles se construyen; límites como profundidad y tamaño de hoja controlan su flexibilidad. Mantendremos el mismo `X_train`, `X_val` y preprocesamiento apropiado para árboles.


```python
bosque = Pipeline([
    ("prep", pre_arbol),
    ("model", RandomForestClassifier(n_estimators=120, max_depth=8,
                                     min_samples_leaf=5, random_state=42, n_jobs=-1)),
])
bosque.fit(X_train, y_train)  # Ningún estadístico se aprende de Validation/Test.
pred_bosque_val = bosque.predict(X_val)
modelos["Random Forest"] = bosque
resultados.append(resumen(y_val, pred_bosque_val, "Random Forest"))
pd.DataFrame(resultados).set_index("modelo").round(3)
```

### Árbol frente a bosque: lee las cuatro columnas

Si el bosque sube Accuracy pero baja Recall de `yes`, puede estar acertando más negativos mientras pierde suscriptores. Si mejora F1, verifica también si Precision o Recall justifican el cambio. Ambos usan idéntica muestra y Validation; así la diferencia no se debe a que uno recibió más contactos. El bosque cuesta más tiempo de entrenamiento y explica peor una decisión individual.


## KNN. K vecinos más cercanos (KNN)

KNN asigna la clase más frecuente entre las observaciones cercanas. Una variable con escala grande dominaría la distancia; por eso StandardScaler se ajusta **solo dentro de Train** en un Pipeline. Un `k` pequeño puede reaccionar al ruido; uno grande suaviza las fronteras y puede ignorar grupos minoritarios.

<p align="center">
    <img src="../material_apoyo/knn.png" width="550">
</p>

## KNN: una predicción por vecindad

Supón que cada observación es un punto en un espacio. Para un cliente nuevo, KNN calcula distancias a los casos de Train, escoge los `k` más cercanos y vota por su clase. Con `k=1`, decide un solo vecino: puede reaccionar al ruido. Con `k` grande, la frontera se suaviza y los negativos, más numerosos, pueden dominar la votación.

En dos variables, la distancia euclidiana es $d(x,z)=\sqrt{(x_1-z_1)^2+(x_2-z_2)^2}$. En cientos de indicadores One-Hot la geometría es menos intuitiva; el principio sigue siendo que las escalas y la representación determinan quién cuenta como «cercano».

```python
# Votación KNN manual en un ejemplo de dos dimensiones
consulta = np.array([2.0, 2.0])
X_juguete = np.array([[1.0, 1.0], [2.5, 2.0], [4.0, 3.0], [1.5, 3.0]])
clases_juguete = np.array([0, 1, 1, 0])
distancias = np.linalg.norm(X_juguete - consulta, axis=1)
orden = np.argsort(distancias)
print("distancias:", np.round(distancias, 2))
print("orden de vecinos:", orden)
print("clases de los 3 vecinos:", clases_juguete[orden[:3]])
print("clase por voto mayoritario:", int(np.bincount(clases_juguete[orden[:3]]).argmax()))
```

### Sigue el cálculo manual

`np.linalg.norm(..., axis=1)` calcula la distancia desde la consulta a cada fila. `argsort` devuelve los índices de menor a mayor distancia; `[:3]` elige tres vecinos. Observa sus etiquetas y cuenta votos. Si uno cambia de coordenada o etiqueta, puede cambiar la decisión: este pequeño ejemplo revela el mecanismo, pero no representa la mezcla real de variables de Bank Marketing.

### Por qué `k` y los pesos importan

Con `weights='distance'`, los vecinos más cercanos pesan más que los que están en el borde del vecindario. Al comparar `k=5,15,31`, mantenemos iguales datos, split, escalado y ponderación; así aislamos parcialmente el efecto de ampliar el vecindario. Aun así, Validation se usa para seleccionar y una sola partición no permite afirmar estabilidad universal.

No interpretes `k=31` como «31 veces más evidencia»: KNN no estima incertidumbre de forma directa y el número de vecinos no sustituye el análisis del costo de error.

### KNN en datos de muchas dimensiones

El ejemplo manual tiene dos variables y se puede dibujar; Bank Marketing, después de One-Hot, tiene muchas más. En dimensiones altas, las distancias pueden volverse menos contrastantes: el vecino más cercano deja de ser tan claramente cercano. Por eso no basta con escalar; también importa cuántas variables incluimos y cuánta señal útil contiene cada una.

Este límite se discute como advertencia, no como razón automática para descartar KNN. Compara su validación y costo computacional en tu conjunto.

```python
# En vivo puedes seguir k=15; las otras opciones quedan explicadas para profundizar.
# Todos conservan X_train/X_val, la clase positiva y el preprocesamiento escalado.
for k in [5, 15, 31]:
    modelo = Pipeline([("prep", pre), ("model", KNeighborsClassifier(n_neighbors=k, weights="distance"))])
    modelo.fit(X_train, y_train)
    modelos[f"KNN k={k}"] = modelo
    resultados.append(resumen(y_val, modelo.predict(X_val), f"KNN k={k}"))
tabla_knn = pd.DataFrame(resultados).set_index("modelo")
tabla_knn.round(3)
```

### Interpreta KNN

Compara Recall y F1, no solo Accuracy. Si al aumentar `k` cae Recall, el vecindario puede estar favoreciendo la clase mayoritaria. Si cambia poco, el resultado puede ser robusto a estos valores de `k`, pero tres configuraciones no bastan para afirmar estabilidad. El escalado protege la distancia; One-Hot hace que las categorías entren como indicadores, una decisión cuyo efecto también debe revisarse.

### Interpreta la tabla de KNN con un orden fijo

1. Compara cada Recall de `yes` con cero del baseline: ¿cuántos positivos aparecen ahora?
2. Mira Precision: ¿cuántas alertas serían correctas?
3. Compara F1 si necesitas equilibrar ambas; revisa Accuracy al final.
4. Observa qué cambia al crecer `k` y si la distancia de Validation entre modelos parece sustantiva.

Si F1 mejora pero Recall sigue bajo, el modelo puede ser útil para priorizar una lista sin resolver el problema completo. Si KNN apenas supera el baseline, no hay obligación de elegirlo: esa es una conclusión válida.

# SVM. Suport Vector Machine

SVM busca una frontera con margen amplio. `C` controla cuánto penaliza errores de Train: valores grandes fuerzan menos errores de entrenamiento y pueden generar fronteras más ajustadas. Kernel lineal representa una frontera lineal; RBF permite fronteras curvas. `gamma` regula el alcance de cada observación en RBF. El escalado es esencial.

<p align="center">
    <img src="../material_apoyo/svm.png" width="650">
</p>

## SVM: separar con un margen

Para una frontera lineal, SVM busca un hiperplano $w^Tx+b=0$ que separe clases y maximice el margen: la distancia entre la frontera y los puntos más cercanos. Esos puntos críticos son los **vectores de soporte**. Un margen amplio suele limitar sensibilidad a cambios pequeños, aunque el resultado depende de la representación y los datos.

En problemas no separables sin errores, SVM permite violaciones. `C` controla el compromiso entre margen amplio y penalización de errores de Train: un `C` mayor penaliza más las violaciones y puede ajustarse más a Train; un `C` menor tolera errores para una frontera más regular. No hay valor universalmente correcto.

### Kernel lineal y kernel RBF

Un kernel permite construir fronteras curvas sin escribir explícitamente todas las combinaciones de variables. `linear` es una frontera lineal en el espacio procesado. `rbf` compara similitud local entre observaciones y puede representar fronteras no lineales.

En RBF, `gamma` controla el alcance de cada observación: gamma alto produce influencia más local y fronteras más onduladas; gamma bajo produce influencia más extendida y fronteras suaves. `gamma='scale'` es una regla automática de partida, no una prueba de que esa configuración sea óptima. Más flexibilidad puede reducir sesgo y elevar varianza.

### El escalado es parte del modelo SVM

Una variable medida en miles puede dominar el producto interno y las distancias del kernel; otra expresada en porcentajes parecería irrelevante por sus unidades. Por ello usamos el mismo preprocesamiento escalado que en KNN, dentro del Pipeline y ajustado solo con Train.

`class_weight='balanced'` repondera las clases de Train según su frecuencia. Es una manera de decirle al ajuste que los errores en `yes` no deben quedar eclipsados por los `no`; **no** aumenta mágicamente la cantidad de suscriptores ni garantiza mejor F1.

### Qué variables delimitan el margen

Los vectores de soporte son observaciones cercanas a la frontera; moverlas puede cambiar el separador. Muchos puntos lejanos pueden influir menos en una SVM lineal una vez fijado el margen. Esto ofrece una intuición geométrica distinta a KNN, que conserva toda la muestra de Train para consultar vecindades.

En la SVM RBF, la frontera deja de ser una recta en el espacio original. Explicar qué variable movió una decisión requiere herramientas adicionales y debe considerar correlación y codificación One-Hot.

```python
# En clase, empieza con SVM lineal; RBF y el segundo C son profundización.
# Se entrena sobre exactamente la misma muestra y división de los otros enfoques.
for kernel, C in [("linear", 1), ("rbf", .5), ("rbf", 2)]:
    modelo = Pipeline([("prep", pre), ("model", SVC(kernel=kernel, C=C, gamma="scale", class_weight="balanced"))])
    modelo.fit(X_train, y_train)
    nombre = f"SVM {kernel}, C={C}"
    modelos[nombre] = modelo
    resultados.append(resumen(y_val, modelo.predict(X_val), nombre))
tabla = pd.DataFrame(resultados).drop_duplicates("modelo").set_index("modelo").sort_values("f1_yes", ascending=False)
tabla.round(3)
```

## Comparación de cuatro enfoques: mismo problema, mismas filas

Para no escoger retrospectivamente entre muchas variantes, mostramos una configuración de **cada enfoque**: árbol, bosque, KNN (`k=15`) y SVM lineal (`C=1`). Todas las filas de la tabla proceden de las **mismas observaciones de Validation**; el baseline se muestra aparte como referencia. Hay preprocesadores distintos por necesidades del algoritmo, pero ninguno ve Validation al ajustarse. Primero declara si tu decisión prioriza F1, Recall o Precision; F1 será solo la regla didáctica por defecto. Si otro costo es relevante, cambia la regla **antes** de abrir Test.


```python
cuatro_enfoques = ["Árbol", "Random Forest", "KNN k=15", "SVM linear, C=1"]
tabla_cuatro = pd.DataFrame([
    resumen(y_val, modelos[nombre].predict(X_val), nombre)
    for nombre in cuatro_enfoques
]).set_index("modelo")
print("Baseline en las mismas filas:")
print(pd.DataFrame([resumen(y_val, pred_base, "Baseline")]).set_index("modelo").round(3))
tabla_cuatro.round(3)
```

### Interpreta kernel, `C` y métricas juntos

Compara primero lineal con RBF; luego compara los dos `C` de RBF. Cambiamos pocos valores para que puedas atribuir tentativamente una diferencia a una decisión concreta. Un F1 mayor en Validation indica mejor equilibrio observado allí, no causalidad ni rendimiento futuro.

Si RBF mejora mucho Train pero no Validation, es señal de que una frontera más compleja no generaliza. Si Recall aumenta mientras Precision cae, el SVM rescata más `yes` a costa de falsas alertas. Traduce esa diferencia a volumen de llamadas o de revisiones antes de preferir un modelo.

### Interpreta la comparación de modelos

Busca el compromiso entre Precision y Recall. `class_weight='balanced'` aumenta el peso de la clase minoritaria, pero no garantiza un resultado superior; puede elevar Recall a costa de más falsos positivos. Kernel RBF con `C` más alto puede ajustar fronteras más complejas, pero decide según Validation. En este diseño SVM se entrena sobre una muestra estratificada para ajustarse al tiempo de clase; reporta que el muestreo limita la comparación con entrenamientos de tamaño completo.

## Umbral: mover la frontera de la decisión

Muchos clasificadores calculan un score continuo y después convierten el score en clase mediante un **umbral**. Bajar el umbral suele etiquetar más observaciones como positivas: normalmente suben los casos positivos detectados (Recall) y también pueden subir los falsos positivos. Subirlo suele hacer las alertas más conservadoras y puede mejorar Precision mientras disminuye Recall.

Este demo usa scores inventados para explicar el mecanismo. No son probabilidades de Bank Marketing ni un umbral recomendado. En producción se elige el umbral con Validation y una función de costo explícita; Test permanece intacto.

```python
y_umbral_demo = np.array([0, 0, 1, 1, 1, 0, 1, 0])
score_demo = np.array([.10, .35, .40, .55, .70, .72, .88, .90])
for umbral in [.35, .50, .75]:
    pred_demo = (score_demo >= umbral).astype(int)
    print("umbral", umbral,
          "Precision", round(precision_score(y_umbral_demo, pred_demo, zero_division=0), 2),
          "Recall", round(recall_score(y_umbral_demo, pred_demo, zero_division=0), 2),
          "F1", round(f1_score(y_umbral_demo, pred_demo, zero_division=0), 2))
```

### Construir una regla de selección defendible

El notebook compara cuatro enfoques predefinidos por F1 de Validation como regla didáctica. Si una institución define que un falso negativo cuesta diez veces más, maximizar F1 podría no ser la regla adecuada. Primero se declara la función de decisión; luego se selecciona modelo/umbral; finalmente se informa el resultado Test.

Si dos modelos empatan, considera simplicidad, estabilidad, tiempo de inferencia, facilidad para explicar errores y mantenimiento. «Mayor métrica» no basta cuando las diferencias son pequeñas.

```python
# La comparación final es entre cuatro enfoques predefinidos, no entre resultados de Test.
# F1 es una regla didáctica: cámbiala antes de evaluar Test si el costo favorece Recall/Precision.
elegido = tabla_cuatro["f1_yes"].idxmax()
modelo_final = modelos[elegido]
print("Elegido usando solo Validation:", elegido)
pred_val = modelo_final.predict(X_val)
ConfusionMatrixDisplay.from_predictions(y_val, pred_val, display_labels=["no", "yes"], values_format="d")
plt.title(f"Matriz de confusión — {elegido} — Validation")
plt.show()
print(classification_report(y_val, pred_val, target_names=["no", "yes"], zero_division=0))
```

### Lee la matriz de confusión

Filas = clase real; columnas = clase predicha. Arriba a la derecha están los falsos positivos; abajo a la izquierda, los falsos negativos. Interpreta conteos absolutos y métricas juntas. Una mejora en Recall puede implicar más clientes contactados en vano; una Precision alta con Recall muy bajo puede dejar muchas suscripciones sin detectar.

### Una matriz se interpreta en términos de casos

Empieza por las celdas, no por el color. Los falsos negativos son personas que contrataron y no detectamos; los falsos positivos son contactos que marcamos y no contrataron. Después conecta esas cuentas con Recall (usa `VP+FN`) y Precision (usa `VP+FP`). El `classification_report` añade *support*, el número real de observaciones por clase.

Una matriz de Validation ayuda a explicar el tipo de fallo. No informa automáticamente cuánto costará ese fallo, si los errores ocurren en un grupo demográfico particular o si el modelo es aceptable legalmente. Esas preguntas requieren diseño y datos adicionales.

`predict()` usa las reglas de clase predeterminadas del modelo escogido; SVM usa su frontera predeterminada. En sistemas con scores probabilísticos se puede elegir otro umbral usando Validation; jamás se ajusta mirando Test. Este ejercicio mantiene el umbral predeterminado para concentrarse en los cuatro clasificadores y sus métricas. El resultado final es una estimación en datos retenidos, no una promesa de desempeño futuro ni una recomendación institucional.

### Validation frente a Test: qué significa que se parezcan

Si las métricas de ambos conjuntos son parecidas, es una señal de consistencia para esta división; no demuestra que una futura campaña conserve la misma distribución. Si Test es peor, puede haber variación muestral, cambio temporal o dependencia de variables específicas. No regreses a cambiar `k`, kernel o umbral mirando repetidamente Test.

Nuestro muestreo de 12 000 casos hace la práctica manejable, pero la estimación no reemplaza una validación temporal ni un piloto prospectivo. La población es portuguesa, las campañas son históricas y la tasa base puede cambiar.

### Cómo cambia la tasa base

Aunque Recall se mantenga, si la proporción real de `yes` cambia en una población nueva, el número de alertas correctas e incorrectas puede cambiar y Precision suele verse afectada. El cambio temporal o poblacional se llama *shift*; las métricas históricas no garantizan el comportamiento bajo ese cambio.

Por eso un sistema aplicado requiere monitorear entradas, calidad de etiquetas y distribución de clases, y volver a validar antes de reutilizar umbrales.
